**เป้าหมาย (target) และประเภทงาน (classification / regression / unsupervised)**

**1. ความสำคัญด้านการบริหารจัดการบริการสาธารณะ:**

*   เรื่องร้องเรียนแต่ละประเภทใช้ระยะเวลาดำเนินการแตกต่างกัน ซึ่งอาจส่งผลต่อความล่าช้าและการสะสมของเรื่องร้องเรียน

*   การคาดการณ์ระยะเวลาปิดเรื่องช่วยให้หน่วยงานวางแผนกำลังคน จัดสรรทรัพยากร และปรับปรุงประสิทธิภาพการดำเนินงานได้

* การวิเคราะห์ความแตกต่างตามพื้นที่และฤดูกาลช่วยให้เข้าใจรูปแบบการจัดการเรื่องร้องเรียนในช่วงเวลาต่าง ๆ

**2. กลุ่มผู้ได้รับประโยชน์โดยตรง (Target Stakeholders):**

*   **หน่วยงาน NYC 311:**
ใช้ผลการทำนายเพื่อประเมินระยะเวลาปิดเรื่องและติดตามเรื่องร้องเรียนที่อาจใช้เวลานาน

*   **หน่วยงานภาครัฐที่รับผิดชอบ (City Agencies):**
ใช้วางแผนบุคลากรและจัดสรรทรัพยากรให้เหมาะสมกับประเภทและจำนวนเรื่องร้องเรียน

*   **ผู้บริหารเมือง (City Planners):** ใช้วิเคราะห์ความแตกต่างของระยะเวลาปิดเรื่องในแต่ละพื้นที่และช่วงฤดูกาล เพื่อวางแผนปรับปรุงบริการสาธารณะ

*   **ประเภทงาน (Task Type):** การถดถอย (Regression Task) เพื่อทำนายระยะเวลาปิดเรื่องร้องเรียน ซึ่งเป็นค่าตัวเลขต่อเนื่อง มีหน่วยเป็นชั่วโมง

*   **ตัวแปรเป้าหมาย (Target Variable):** resolution_time คือ จำนวนชั่วโมงตั้งแต่วันที่รับแจ้งเรื่อง (created_date) จนถึงวันที่ปิดเรื่อง (closed_date) คำนวณจากผลต่างของเวลาในหน่วยวินาทีหารด้วย 3,600 โดยใช้เฉพาะรายการที่มีวันที่ปิดเรื่องถูกต้อง

*   **ตัวแปรที่ใช้ในการทำนาย (Features):** ประเภทเรื่องร้องเรียน (complaint_type) หน่วยงาน (agency) เขตพื้นที่ (borough) และข้อมูลวันที่และเวลารับแจ้ง (created_date) ซึ่งสามารถนำมาสร้างตัวแปรชั่วโมง วันในสัปดาห์ เดือน และฤดูกาลได้

*   **คุณลักษณะพิเศษของปัญหา:** ระยะเวลาปิดเรื่องอาจมีการกระจายแบบเบ้ขวา (Right-skewed Distribution) เนื่องจากบางเรื่องอาจใช้เวลาดำเนินการนานกว่าปกติ จึงต้องตรวจสอบการกระจายของข้อมูลและค่าผิดปกติก่อนสร้างโมเดล รวมถึงป้องกัน Data Leakage โดยใช้เฉพาะข้อมูลที่ทราบ ณ เวลารับแจ้งเรื่อง

**แหล่งอ้างอิง:** [NYC Open Data – 311 Service Requests from 2020 to Present](https://data.cityofnewyork.us/Social-Services/311-Service-Requests-from-2020-to-Present/erm2-nwe9/about_data)

## 3. Data Loading

```python
import os
import sys
import time
import requests
import pandas as pd
import numpy as np

# สร้างโฟลเดอร์เก็บข้อมูลดิบ
data_dir = "data"
os.makedirs(data_dir, exist_ok=True)

# ลิงก์ SODA API ทางการของ NYC Open Data (311 Service Requests)
BASE_URL = "https://data.cityofnewyork.us/resource/erm2-nwe9.csv"

# คอลัมน์ที่จำเป็นต่อการทำนายและวิเคราะห์ ML
columns = [
    "unique_key", "created_date", "closed_date",
    "complaint_type", "agency", "borough",
    "incident_zip", "status"
]

PAGE_SIZE = 50000
session = requests.Session()

print("--- เริ่มดาวน์โหลดข้อมูล NYC 311 รายเดือน (ปี 2025) ---")
t_dl_start = time.time()
month_file_paths = []

# 1. ดาวน์โหลดข้อมูลแยกรายเดือนครบทั้ง 12 เดือน (ปี 2025)
for month in range(1, 13):
    filename = os.path.join(data_dir, f"nyc311_2025_{month:02d}.csv")
    month_file_paths.append(filename)

    # ข้ามการดาวน์โหลดถ้าไฟล์เดิมมีขนาดสมบูรณ์แล้ว
    if os.path.exists(filename) and os.path.getsize(filename) > 100 * 1024:
        print(f"เดือน {month:02d}: พบไฟล์เดิมเรียบร้อยแล้ว ({os.path.getsize(filename)/(1024**2):.1f} MB)")
        continue

    start = f"2025-{month:02d}-01T00:00:00"
    end = f"2025-{month+1:02d}-01T00:00:00" if month < 12 else "2026-01-01T00:00:00"

    offset = 0
    first_page = True
    month_rows = 0
    temp_file = filename + ".part"

    if os.path.exists(temp_file):
        os.remove(temp_file)

    while True:
        params = {
            "$select": ",".join(columns),
            "$where": f"created_date >= '{start}' AND created_date < '{end}'",
            "$order": "created_date ASC, unique_key ASC",
            "$limit": PAGE_SIZE,
            "$offset": offset
        }

        # Retry mechanism สำหรับการดาวน์โหลดบน Colab
        for attempt in range(4):
            try:
                response = session.get(BASE_URL, params=params, timeout=120)
                response.raise_for_status()
                break
            except requests.RequestException:
                if attempt == 3:
                    raise
                time.sleep(4 * (attempt + 1))

        from io import BytesIO
        chunk = pd.read_csv(BytesIO(response.content), dtype="string")
        n = len(chunk)

        if n == 0:
            break

        chunk.to_csv(temp_file, mode="w" if first_page else "a", header=first_page, index=False)
        first_page = False
        month_rows += n
        offset += n

        if n < PAGE_SIZE:
            break

    if first_page:
        pd.DataFrame(columns=columns).to_csv(temp_file, index=False)

    os.replace(temp_file, filename)
    print(f"เดือน {month:02d}: ดาวน์โหลดสำเร็จ {month_rows:,} แถว")

session.close()
print(f"ดาวน์โหลดข้อมูลรายเดือนเสร็จสิ้นใน {(time.time() - t_dl_start)/60:.1f} นาที")

# 2. อ่านไฟล์รายเดือนและรวมข้อมูล (Memory-Safe Loading)
print("\n--- เริ่มรวมข้อมูลรายเดือนและประมวลผล Memory Usage ---")
t_load_start = time.time()

loaded_dfs = []
total_raw_rows = 0

for file_path in month_file_paths:
    # โหลดไฟล์แบบ string แล้ว optimize memory ต่อเนื่อง
    df_chunk = pd.read_csv(file_path, dtype="string")
    total_raw_rows += len(df_chunk)
    loaded_dfs.append(df_chunk)

# รวม DataFrame ทั้ง 12 เดือนเข้าด้วยกัน
raw_df = pd.concat(loaded_dfs, ignore_index=True)
del loaded_dfs  # ลบลิสต์ชั่วคราวเพื่อคืนพื้นที่ RAM

# แปลง Data Types เพื่อประหยัดพื้นที่ RAM
raw_df["unique_key"] = raw_df["unique_key"].astype("string")
raw_df["complaint_type"] = raw_df["complaint_type"].astype("category")
raw_df["agency"] = raw_df["agency"].astype("category")
raw_df["borough"] = raw_df["borough"].astype("category")
raw_df["status"] = raw_df["status"].astype("category")

load_time = time.time() - t_load_start

print(f"\nรวมข้อมูลเสร็จสิ้นใน {load_time:.2f} วินาที")
print(f"จำนวนแถวดิบรวมทั้ง 12 เดือน: {len(raw_df):,} แถว")
print(f"ขนาดหน่วยความจำของข้อมูลดิบใน RAM: {raw_df.memory_usage(deep=True).sum() / (1024**2):.2f} MB")

# ตรวจสอบเกณฑ์ขั้นต่ำ 1,000,000 แถว
if len(raw_df) >= 1_000_000:
    print("ผ่านเกณฑ์ข้อมูลอย่างน้อย 1,000,000 แถวสำหรับโปรเจกต์")
else:
    print("ไม่ผ่านเกณฑ์ข้อมูลขั้นต่ำ")

# บันทึกเป็นไฟล์ดิบรวมนอกโฟลเดอร์ data/ เพื่อป้องกันไฟล์ซ้ำซ้อน
raw_df.to_csv("nyc311_2025_raw.csv", index=False)
print("บันทึกไฟล์รวม 'nyc311_2025_raw.csv' เรียบร้อยแล้ว")

# แสดงตัวอย่างข้อมูล
display(raw_df.head(5))
```
