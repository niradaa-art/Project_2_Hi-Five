## 1. ปัญหาและคำถามที่สนใจ

### ปัญหา/คำถามที่สนใจคืออะไร

นครนิวยอร์ก (New York City) มีระบบ NYC 311 สำหรับรับแจ้งปัญหาและเรื่องร้องเรียนจากประชาชน เช่น เสียงรบกวน ขยะตกค้าง ถนนชำรุด และปัญหาที่อยู่อาศัย โดยเรื่องร้องเรียนแต่ละประเภทอาจใช้ระยะเวลาในการปิดเรื่องแตกต่างกัน ขึ้นอยู่กับประเภทปัญหา หน่วยงานที่รับผิดชอบ พื้นที่ และช่วงเวลาที่ได้รับแจ้ง รวมถึงฤดูกาลที่อาจเกี่ยวข้องกับลักษณะของเรื่องร้องเรียน การคาดการณ์ระยะเวลาปิดเรื่องจึงมีประโยชน์ต่อการวางแผนและจัดสรรทรัพยากรของหน่วยงานภาครัฐ

**คำถามหลักของโครงงาน:**

“ข้อมูลเรื่องร้องเรียนของประชาชนในนครนิวยอร์ก (NYC 311 Service Requests) สามารถนำมาสร้างโมเดล Machine Learning เพื่อคาดการณ์ระยะเวลาในการปิดเรื่องร้องเรียน (`resolution_time`) ได้แม่นยำเพียงใด และปัจจัยต่าง ๆ เช่น ประเภทเรื่องร้องเรียน หน่วยงานที่รับผิดชอบ เขตพื้นที่ (`borough`) วันที่และเวลาที่แจ้งเรื่อง รวมถึงฤดูกาล มีความสำคัญต่อผลการทำนายมากน้อยเพียงใด?”

### ปัญหา/คำถามนี้มีความสำคัญอย่างไร และใครจะนำผลไปใช้

**1. ความสำคัญด้านการบริหารจัดการบริการสาธารณะ**

- เรื่องร้องเรียนแต่ละประเภทใช้ระยะเวลาดำเนินการแตกต่างกัน ซึ่งอาจทำให้เกิดความล่าช้าและการสะสมของเรื่องร้องเรียนที่ยังไม่ปิด
- การคาดการณ์ระยะเวลาปิดเรื่องช่วยสนับสนุนการวางแผนกำลังคน จัดสรรทรัพยากร และปรับปรุงประสิทธิภาพการดำเนินงาน
- การวิเคราะห์ความแตกต่างตามพื้นที่และฤดูกาลช่วยให้เข้าใจรูปแบบการจัดการเรื่องร้องเรียนในช่วงเวลาต่าง ๆ

**2. กลุ่มผู้ได้รับประโยชน์ (Target Stakeholders)**

- **หน่วยงาน NYC 311:** ใช้ผลการทำนายเพื่อประเมินระยะเวลาปิดเรื่องและติดตามเรื่องร้องเรียนที่อาจใช้เวลานาน
- **หน่วยงานภาครัฐที่รับผิดชอบ (City Agencies):** ใช้วางแผนบุคลากรและจัดสรรทรัพยากรให้เหมาะสมกับประเภทและจำนวนเรื่องร้องเรียน
- **ผู้บริหารเมือง (City Planners):** ใช้วิเคราะห์ความแตกต่างของระยะเวลาปิดเรื่องในแต่ละพื้นที่และช่วงฤดูกาล เพื่อประกอบการปรับปรุงบริการสาธารณะ
- **ประชาชน (Residents):** สามารถทราบระยะเวลาปิดเรื่องโดยประมาณ หากมีการนำโมเดลไปใช้งานจริง

### เป้าหมาย (Target) และประเภทงาน (Classification / Regression / Unsupervised)

- **ประเภทงาน (Task Type):** Regression เป็นการทำนายค่าตัวเลขต่อเนื่อง โดยโครงงานนี้ต้องการทำนายจำนวนชั่วโมงที่ใช้ในการปิดเรื่องร้องเรียน

- **ตัวแปรเป้าหมาย (Target Variable):** `resolution_time` คือ ระยะเวลาตั้งแต่วันที่รับแจ้งเรื่อง (`created_date`) จนถึงวันที่ปิดเรื่อง (`closed_date`) มีหน่วยเป็นชั่วโมง โดยคำนวณจากผลต่างของเวลาในหน่วยวินาทีหารด้วย 3,600 และใช้เฉพาะรายการที่มีวันที่ถูกต้อง

- **ตัวแปรที่ใช้ในการทำนาย (Features):** ประเภทเรื่องร้องเรียน (`complaint_type`) หน่วยงานที่รับผิดชอบ (`agency`) เขตพื้นที่ (`borough`) และวันที่และเวลาที่แจ้งเรื่อง (`created_date`) ซึ่งสามารถนำมาสร้างตัวแปรชั่วโมง วันในสัปดาห์ เดือน และฤดูกาล โดยต้องตรวจสอบว่าตัวแปรเหล่านี้ทราบได้ ณ เวลารับแจ้งจริง

- **คุณลักษณะพิเศษของปัญหา:** ระยะเวลาปิดเรื่องอาจมีการกระจายแบบเบ้ขวา (Right-skewed Distribution) เนื่องจากบางเรื่องอาจใช้เวลาดำเนินการนานกว่าปกติ จึงต้องตรวจสอบการกระจายของข้อมูลและค่าผิดปกติก่อนสร้างโมเดล รวมถึงป้องกัน Data Leakage โดยไม่ใช้ข้อมูลที่เกิดขึ้นภายหลังการรับแจ้งเรื่องเป็น Features

### 2. จำนวนแถว จำนวนคอลัมน์ และ Data Dictionary

จากการโหลดข้อมูล NYC 311 Service Requests ปี 2025 ผ่าน NYC Open Data API ได้ผลดังนี้

- **จำนวนแถว (Rows):** ประมาณ 3.65 ล้านแถว (ตัวเลขที่แน่นอนจากการรันแสดงในข้อ 3)
- **จำนวนคอลัมน์ (Columns):** 8 คอลัมน์
- **Memory Usage:** ประมาณ 0.9 GB (ตัวเลขที่แน่นอนแสดงในข้อ 3)
- **จำนวนไฟล์ที่นำมารวม:** 12 ไฟล์ (แยกตามเดือนมกราคม–ธันวาคม 2025)
- **วิธีรวมข้อมูล:** ใช้ `pd.concat()` เพื่อรวมข้อมูลรายเดือนเป็นชุดเดียว โดยไม่จำเป็นต้อง Merge กับตารางอื่น

**หมายเหตุเรื่องการเลือกคอลัมน์และวันที่ดึงข้อมูล**

- ตารางต้นฉบับบน NYC Open Data มีคอลัมน์จำนวนมาก (ดูรายการเต็มได้ที่หน้า About Data) กลุ่มเลือกดึงเพียง 8 คอลัมน์ผ่าน `$select` เพราะเป็นคอลัมน์ที่จำเป็นต่อการสร้าง Target (`created_date`, `closed_date`), ต่อ Features ที่ทราบ ณ เวลารับแจ้ง (`complaint_type`, `agency`, `borough`) และต่อการตรวจคุณภาพ/ตรวจ leakage (`unique_key`, `incident_zip`, `status`) โดย `closed_date` ใช้สร้าง Target และ `status` ใช้ตรวจคุณภาพเท่านั้น ไม่ใช่ Features
- ค่า `closed_date` เป็นค่า ณ วันที่ดึงข้อมูล (snapshot) เรื่องที่ยังไม่ปิด ณ วันนั้นจะไม่มีวันปิด จึงต้องระบุวัน snapshot ไว้ (บันทึกอัตโนมัติในข้อ 3 เป็นตัวแปร `SNAPSHOT_DATE`)

## 3. Data Loading

```python
import os
import time
import gc
import requests
import pandas as pd
from io import BytesIO

# 1. กำหนดแหล่งข้อมูลและตัวแปร
BASE_URL = "https://data.cityofnewyork.us/resource/erm2-nwe9.csv"
COUNT_URL = "https://data.cityofnewyork.us/resource/erm2-nwe9.json"

data_dir = "data"
os.makedirs(data_dir, exist_ok=True)

YEAR = 2025
PAGE_SIZE = 50000

columns = [
    "unique_key", "created_date", "closed_date",
    "complaint_type", "agency", "borough",
    "incident_zip", "status"
]

session = requests.Session()
t_start = time.time()

# ฟังก์ชันเรียก API พร้อมลองใหม่เมื่อเชื่อมต่อขัดข้อง
def get_response(url, params):
    for attempt in range(4):
        try:
            response = session.get(
                url, params=params, timeout=120
            )
            response.raise_for_status()
            return response
        except requests.RequestException:
            if attempt == 3:
                raise
            time.sleep(5 * (attempt + 1))

# ฟังก์ชันนับจำนวนแถวใน CSV
def count_rows(path):
    return sum(
        len(chunk)
        for chunk in pd.read_csv(
            path, chunksize=50000, dtype="string"
        )
    )

# 2. ดาวน์โหลดข้อมูลปี 2025 แยกตามเดือน
month_files = []
monthly_counts = []

print("===== Download NYC 311 Data =====")

try:
    for month in range(1, 13):

        start = f"{YEAR}-{month:02d}-01T00:00:00"
        end = (
            f"{YEAR}-{month+1:02d}-01T00:00:00"
            if month < 12 else
            f"{YEAR+1}-01-01T00:00:00"
        )

        where = (
            f"created_date >= '{start}' "
            f"AND created_date < '{end}'"
        )

        filename = os.path.join(
            data_dir, f"nyc311_{YEAR}_{month:02d}.csv"
        )

        # ตรวจจำนวนข้อมูลของเดือนจาก API
        result = get_response(COUNT_URL, {
            "$select": "count(*) as total",
            "$where": where
        }).json()

        expected_rows = int(result[0]["total"])

        # ตรวจไฟล์เดิมก่อนดาวน์โหลด
        use_existing = False

        if os.path.exists(filename):
            try:
                existing_rows = count_rows(filename)

                if existing_rows == expected_rows:
                    use_existing = True
                    print(
                        f"เดือน {month:02d}: "
                        f"ใช้ไฟล์เดิม {existing_rows:,} แถว"
                    )
            except Exception:
                use_existing = False

        if not use_existing:

            temp_file = filename + ".part"
            if os.path.exists(temp_file):
                os.remove(temp_file)

            offset = 0
            first_page = True

            while offset < expected_rows:

                limit = min(
                    PAGE_SIZE, expected_rows - offset
                )

                params = {
                    "$select": ",".join(columns),
                    "$where": where,
                    "$order": "created_date ASC, unique_key ASC",
                    "$limit": limit,
                    "$offset": offset
                }

                response = get_response(BASE_URL, params)
                chunk = pd.read_csv(
                    BytesIO(response.content),
                    dtype="string"
                )

                if len(chunk) == 0:
                    raise RuntimeError(
                        f"เดือน {month:02d}: API ส่งข้อมูลไม่ครบ"
                    )

                chunk = chunk.reindex(columns=columns)

                chunk.to_csv(
                    temp_file,
                    mode="w" if first_page else "a",
                    header=first_page,
                    index=False
                )

                offset += len(chunk)
                first_page = False

                del chunk

            # ตรวจไฟล์ที่ดาวน์โหลดเสร็จ
            if count_rows(temp_file) != expected_rows:
                raise ValueError(
                    f"เดือน {month:02d}: จำนวนแถวไม่ตรง"
                )

            os.replace(temp_file, filename)

            print(
                f"เดือน {month:02d}: "
                f"ดาวน์โหลด {expected_rows:,} แถว"
            )

        month_files.append(filename)
        monthly_counts.append(expected_rows)

finally:
    session.close()

# 3. รวมข้อมูลรายเดือน
print("\n===== Combine Data =====")

loaded_dfs = []

for file in month_files:
    part = pd.read_csv(file, dtype="string")
    loaded_dfs.append(part)

rows_before = sum(len(part) for part in loaded_dfs)

raw_df = pd.concat(
    loaded_dfs, ignore_index=True
)

rows_after = len(raw_df)

del loaded_dfs
gc.collect()

print("จำนวนไฟล์:", len(month_files))
print(f"จำนวนแถวก่อนรวม: {rows_before:,}")
print(f"จำนวนแถวหลังรวม: {rows_after:,}")

assert len(month_files) == 12
assert rows_before == rows_after
assert rows_after == sum(monthly_counts)

print("จำนวนแถวก่อนและหลังรวมตรงกัน")
print("ไม่จำเป็นต้อง Merge กับตารางอื่น")

# 4. ปรับชนิดข้อมูลบางคอลัมน์เพื่อประหยัด RAM
for col in [
    "complaint_type", "agency", "borough", "status"
]:
    raw_df[col] = raw_df[col].astype("category")

# 5. ตรวจสอบขนาดและหน่วยความจำ
print("\n===== Dataset Information =====")

print("Data Shape:", raw_df.shape)

memory_mb = (
    raw_df.memory_usage(deep=True).sum() / 1024**2
)

print(f"Memory Usage: {memory_mb:,.2f} MB")

print("\nData Types")
display(raw_df.dtypes.to_frame("Data Type"))

print("\nตัวอย่างข้อมูล")
display(raw_df.head())

# 6. ตรวจสอบข้อมูลเบื้องต้น
print("\n===== Data Validation =====")

print("Unique Key ซ้ำ:",
      f"{raw_df['unique_key'].duplicated().sum():,}")

print("\nMissing Values")
missing = pd.DataFrame({
    "Missing Count": raw_df.isna().sum(),
    "Missing (%)": (
        raw_df.isna().mean() * 100
    ).round(2)
})
display(missing)

# ตรวจสอบวันที่ โดยไม่แก้ข้อมูลดิบ
created = pd.to_datetime(
    raw_df["created_date"], errors="coerce"
)

closed = pd.to_datetime(
    raw_df["closed_date"], errors="coerce"
)

print("\nช่วงวันที่รับแจ้ง")
print("เริ่มต้น:", created.min())
print("สิ้นสุด:", created.max())

# ตรวจสอบความพร้อมของ Target
valid_target = (
    created.notna()
    & closed.notna()
    & (closed >= created)
)

print("\nTarget Availability")
print(
    "คำนวณ resolution_time ได้:",
    f"{valid_target.sum():,}"
)
print(
    "ยังคำนวณไม่ได้:",
    f"{(~valid_target).sum():,}"
)

# ตรวจสอบข้อมูลปี 2025
valid_year = created.dt.year.eq(YEAR)

print("\nข้อมูลปี 2025:",
      f"{valid_year.sum():,}")

assert valid_year.all(), (
    "พบข้อมูลนอกช่วงปี 2025 หรือวันที่ไม่ถูกต้อง"
)

# 7. ตรวจสอบเกณฑ์อาจารย์
print("\n===== Minimum Data Requirement =====")

if len(raw_df) >= 1_000_000:
    print("ผ่านเกณฑ์ข้อมูลอย่างน้อย 1 ล้านแถว")
else:
    print("ยังไม่ผ่านเกณฑ์ข้อมูลขั้นต่ำ")

# 8. บันทึกข้อมูลสำหรับส่งต่อ
output_file = "nyc311_2025_raw.csv"

raw_df.to_csv(
    output_file,
    index=False,
    encoding="utf-8-sig"
)

print("\n===== Final Summary =====")
print("Dataset: NYC 311 Service Requests")
print("Year:", YEAR)
print(f"Rows: {len(raw_df):,}")
print("Columns:", raw_df.shape[1])
print(f"Memory: {memory_mb:,.2f} MB")
print("Combine Method: pd.concat()")
print("Merge: Not required")
print("Saved File:", output_file)
print(
    f"Time: {(time.time()-t_start)/60:.2f} minutes"
)

# 9. บันทึกวัน snapshot ของข้อมูล (ใช้ตีความ Target: "ปิดแล้ว ณ วันที่ดึงข้อมูล")
SNAPSHOT_DATE = pd.Timestamp(max(os.path.getmtime(f) for f in month_files), unit="s")
print("วัน snapshot (เวลาของไฟล์รายเดือนล่าสุด):", SNAPSHOT_DATE.date())
```

## 4. Data Quality Assessment
ตรวจและรายงานเป็นตัวเลข: ชนิดข้อมูล, missing values, ข้อมูลซ้ำ, ค่าผิดปกติ/เป็นไปไม่ได้, ความไม่สอดคล้องของข้อมูล

```python

import pandas as pd
import numpy as np

# เก็บข้อมูลดิบไว้ ไม่แก้ไข raw_df
quality_df = raw_df.copy()
N = len(quality_df)

print("=" * 70)
print("4. DATA QUALITY ASSESSMENT: NYC 311 (2025)")
print("=" * 70)
print(f"จำนวนแถวดิบ: {N:,}")
print(f"จำนวนคอลัมน์: {quality_df.shape[1]}")

# --------------------------------------------------
# 1. Data Types and Memory Usage
# --------------------------------------------------
print("\n=== 1. Data Types and Memory Usage ===")

memory_report = pd.DataFrame({
    "Data Type": quality_df.dtypes.astype(str),
    "Non-Null Count": quality_df.notna().sum(),
    "Missing Count": quality_df.isna().sum(),
    "Memory (MB)": (
        quality_df.memory_usage(deep=True, index=False)
        / (1024 ** 2)
    )
})

display(memory_report.round(3))

print(
    "Total memory (MB):",
    round(quality_df.memory_usage(deep=True).sum() / (1024**2), 2)
)

# --------------------------------------------------
# 2. Missing Values
# --------------------------------------------------
print("\n=== 2. Missing Values Analysis ===")

# รวมทั้ง NA และข้อความว่าง/ช่องว่าง
missing_mask = quality_df.isna().copy()

for col in quality_df.columns:
    if (
        pd.api.types.is_string_dtype(quality_df[col])
        or quality_df[col].dtype == object
    ):
        missing_mask[col] = (
            missing_mask[col]
            | quality_df[col].astype("string").str.strip().eq("").fillna(False)
        )

missing_count = missing_mask.sum()
missing_pct = (missing_count / N * 100) if N else missing_count

missing_report = pd.DataFrame({
    "Missing Rows": missing_count,
    "Missing Percentage (%)": missing_pct.round(4),
    "Non-Missing Rows": N - missing_count
}).sort_values("Missing Rows", ascending=False)

display(missing_report)

# --------------------------------------------------
# 3. Duplicate Rows and Duplicate Keys
# --------------------------------------------------
print("\n=== 3. Duplicate Rows Analysis ===")

exact_dup = quality_df.duplicated().sum()

key_as_string = quality_df["unique_key"].astype("string").str.strip()
key_missing = key_as_string.isna() | key_as_string.eq("").fillna(False)
duplicate_key_mask = (
    key_as_string.duplicated(keep=False) & ~key_missing
)

print(
    f"Exact duplicate rows: {exact_dup:,} "
    f"({exact_dup / N * 100:.4f}%)"
)
print(
    f"Rows involved in repeated unique_key: "
    f"{duplicate_key_mask.sum():,} "
    f"({duplicate_key_mask.sum() / N * 100:.4f}%)"
)
print(f"Missing unique_key: {key_missing.sum():,}")

# --------------------------------------------------
# 4. Invalid / Impossible Values
# --------------------------------------------------
print("\n=== 4. Invalid / Impossible Values ===")

created = pd.to_datetime(
    quality_df["created_date"], errors="coerce"
)
closed = pd.to_datetime(
    quality_df["closed_date"], errors="coerce"
)

# 4.1 Created date invalid / outside year 2025
created_invalid = created.isna()
created_outside_2025 = (
    created.notna()
    & ~created.ge(pd.Timestamp("2025-01-01"))
        .mul(created.lt(pd.Timestamp("2026-01-01")))
)

# 4.2 Closed date provided but invalid
closed_invalid = (
    quality_df["closed_date"].notna()
    & closed.isna()
)

# Treat whitespace-only closed_date as missing, not malformed
closed_text = quality_df["closed_date"].astype("string").str.strip()
closed_blank = closed_text.isna() | closed_text.eq("").fillna(False)
closed_invalid = closed_invalid & ~closed_blank

# 4.3 Resolution time
resolution_time = (closed - created).dt.total_seconds() / 3600

negative_duration = resolution_time.notna() & (resolution_time < 0)
zero_duration = resolution_time.notna() & (resolution_time == 0)

# 4.4 Missing / blank categorical fields
category_cols = ["complaint_type", "agency", "borough", "status"]
category_missing = {}

for col in category_cols:
    s = quality_df[col].astype("string").str.strip()
    category_missing[col] = (s.isna() | s.eq("").fillna(False)).sum()

print(
    f"Invalid created_date: {created_invalid.sum():,} "
    f"({created_invalid.mean() * 100:.4f}%)"
)
print(
    f"Created date outside 2025: {created_outside_2025.sum():,} "
    f"({created_outside_2025.sum() / N * 100:.4f}%)"
)
print(
    f"Invalid nonblank closed_date: {closed_invalid.sum():,} "
    f"({closed_invalid.sum() / N * 100:.4f}%)"
)
print(
    f"Negative resolution time: {negative_duration.sum():,} "
    f"({negative_duration.sum() / N * 100:.4f}%)"
)
print(
    f"Zero resolution time: {zero_duration.sum():,} "
    f"({zero_duration.sum() / N * 100:.4f}%)"
)

print("\nMissing/blank categorical values:")
display(pd.DataFrame({
    "Missing Rows": pd.Series(category_missing),
    "Missing Percentage (%)": (
        pd.Series(category_missing) / N * 100
    ).round(4)
}))

# --------------------------------------------------
# 5. Inconsistent Categorical Values
# --------------------------------------------------
print("\n=== 5. Categorical Consistency ===")

for col in category_cols:
    s = quality_df[col].astype("string").str.strip()
    s = s.mask(s.eq("").fillna(False), pd.NA)

    print(f"\n{col}: {s.nunique(dropna=True):,} distinct nonblank values")
    display(
        s.value_counts(dropna=False).head(15).to_frame("Count")
    )

# รายงานค่าพื้นที่ที่พบ เพื่อให้ตรวจสอบด้วย Data Dictionary
print("\nBorough values:")
display(
    quality_df["borough"]
    .astype("string").str.strip()
    .value_counts(dropna=False)
    .to_frame("Count")
)

# --------------------------------------------------
# Outlier assessment for resolution_time
# --------------------------------------------------
print("\n=== Additional: Resolution Time Outliers ===")

valid_duration = resolution_time[
    resolution_time.notna() & (resolution_time >= 0)
]

if not valid_duration.empty:
    q1 = valid_duration.quantile(0.25)
    q3 = valid_duration.quantile(0.75)
    iqr = q3 - q1
    upper = q3 + 1.5 * iqr

    outlier_mask = valid_duration > upper

    print(f"Valid duration values: {len(valid_duration):,}")
    print(f"Q1: {q1:.3f} hours")
    print(f"Median: {valid_duration.median():.3f} hours")
    print(f"Q3: {q3:.3f} hours")
    print(f"IQR upper bound: {upper:.3f} hours")
    print(f"High-duration outliers: {outlier_mask.sum():,}")
    print(
        f"Outlier percentage: "
        f"{outlier_mask.mean() * 100:.4f}%"
    )

    display(
        valid_duration.describe(
            percentiles=[0.5, 0.75, 0.9, 0.95, 0.99]
        ).to_frame("Resolution Time (Hours)")
    )

# --------------------------------------------------
# Additional: ความไม่สอดคล้องของข้อมูล (status vs closed_date, รูปแบบ zip)
# --------------------------------------------------
print("\n=== Additional: status vs closed_date consistency ===")
has_closed = quality_df["closed_date"].notna().rename("has_closed_date")
status_s = quality_df["status"].astype("string")
display(pd.crosstab(status_s, has_closed))
closed_no_date = int(((status_s == "Closed") & ~has_closed).sum())
open_with_date = int(((status_s != "Closed") & has_closed).sum())
print(f"status = Closed แต่ไม่มี closed_date: {closed_no_date:,}")
print(f"status != Closed แต่มี closed_date: {open_with_date:,}")

print("\n=== Additional: incident_zip format ===")
zip_s = quality_df["incident_zip"].astype("string").str.strip()
zip_ok = zip_s.str.fullmatch(r"\d{5}(-\d{4})?").fillna(False).astype(bool)
bad_zip = int((zip_s.notna() & ~zip_ok).sum())
print(f"incident_zip ที่รูปแบบไม่ใช่ 5 หลัก (หรือ ZIP+4): {bad_zip:,}")
```

## 5. Data Cleaning

ดำเนินการทำความสะอาดข้อมูลตามผลการประเมินคุณภาพข้อมูลในหัวข้อ 4 โดยกำหนดขั้นตอนการจัดการข้อมูลอย่างเป็นลำดับ พร้อมบันทึกจำนวนแถวก่อนดำเนินการ จำนวนแถวที่ถูกนำออก จำนวนแถวคงเหลือ และร้อยละของข้อมูลที่คงเหลือเมื่อเทียบกับข้อมูลดิบเริ่มต้น

### 5.1 ขั้นตอนการทำความสะอาดข้อมูล

| ขั้นตอน | สิ่งที่ดำเนินการ | เหตุผล |
|---|---|---|
| 0. ข้อมูลดิบเริ่มต้น | ใช้ข้อมูล NYC 311 ปี 2025 จำนวน 3,655,042 แถว | เป็นข้อมูลตั้งต้นสำหรับเปรียบเทียบผลก่อนและหลังทำความสะอาด |
| 1. กำจัดข้อมูลซ้ำ | ลบแถวที่ซ้ำกันทุกคอลัมน์ | ป้องกันการนับข้อมูลรายการเดิมซ้ำ |
| 2. ตรวจสอบรหัสรายการ | ตรวจสอบ `unique_key` ที่ซ้ำกันและพิจารณารายละเอียดเพิ่มเติม | ป้องกันการลบข้อมูลที่มีรายละเอียดแตกต่างกันโดยไม่จำเป็น |
| 3. ตรวจสอบวันที่รับแจ้ง | คัดเฉพาะรายการที่ `created_date` ถูกต้องและอยู่ในปี 2025 | ให้ข้อมูลตรงกับช่วงเวลาที่กำหนดในการศึกษา |
| 4. ตรวจสอบวันที่ปิดเรื่อง | คัดเฉพาะรายการที่มี `closed_date` ถูกต้องและสามารถนำมาคำนวณระยะเวลาได้ | จำเป็นต้องมี Target สำหรับการเรียนรู้แบบ Regression |
| 5. คำนวณระยะเวลาปิดเรื่อง | สร้าง `resolution_time` จากผลต่างระหว่าง `closed_date` และ `created_date` โดยแปลงเป็นชั่วโมง | สร้างตัวแปรเป้าหมายสำหรับการทำนาย |
| 6. กำจัดระยะเวลาที่ติดลบ | ตัดรายการที่ `resolution_time` น้อยกว่า 0 | ป้องกันข้อมูลที่ขัดแย้งกับลำดับเวลาที่ควรเป็น |
| 7. จัดการค่าสูญหายของตัวแปรทำนาย | แทนค่าว่างในตัวแปรเชิงหมวดหมู่ที่เลือกใช้ด้วย `Unknown` | รักษารายการที่ยังมีข้อมูลส่วนอื่นเป็นประโยชน์ต่อการวิเคราะห์ |
| 8. สร้างตัวแปรเวลา | สร้าง `hour`, `day_of_week`, `month` และ `season` จาก `created_date` | ใช้วิเคราะห์รูปแบบการรับแจ้งตามเวลาและฤดูกาล |
| 9. ตรวจสอบค่าผิดปกติ | ตรวจสอบค่าระยะเวลาปิดเรื่องที่สูงผิดปกติด้วย IQR | ประเมินความสมเหตุสมผลของข้อมูลก่อนสร้างโมเดล โดยไม่ลบค่าที่อาจเกิดขึ้นจริงโดยอัตโนมัติ |




```python
# Handle duplicates, invalid values, outliers, inconsistent categories

# ==========================================
# 5. DATA CLEANING PIPELINE
# ==========================================

clean_df = raw_df.copy()
initial_rows = len(clean_df)
cleaning_log = []

def record_step(step, action, before, after, reason):
    cleaning_log.append({
        "Step": step,
        "Action": action,
        "Rows Before": before,
        "Rows Removed": before - after,
        "Rows Remaining": after,
        "Remaining (%)": round(after / initial_rows * 100, 4),
        "Reason": reason
    })

print(f"Initial rows: {initial_rows:,}")

# Step 1: Remove exact duplicates
before = len(clean_df)
clean_df = clean_df.drop_duplicates().copy()

record_step(
    1, "Remove exact duplicate rows", before, len(clean_df),
    "ลบเฉพาะแถวที่ซ้ำกันทุกคอลัมน์"
)

# Step 2: Inspect duplicate unique_key
# ไม่ลบ duplicate keys อัตโนมัติ เพราะอาจมีข้อมูลขัดแย้งกัน
key = clean_df["unique_key"].astype("string").str.strip()
missing_key = key.isna() | key.eq("").fillna(False)

dup_key_mask = key.duplicated(keep=False) & ~missing_key

print(
    "Rows involved in duplicate unique_key after exact dedup:",
    f"{dup_key_mask.sum():,}"
)

if dup_key_mask.any():
    display(
        clean_df.loc[dup_key_mask]
        .sort_values("unique_key")
        .head(20)
    )

# Step 3: Parse datetime and retain valid 2025 created_date
before = len(clean_df)

clean_df["created_dt"] = pd.to_datetime(
    clean_df["created_date"], errors="coerce"
)
clean_df["closed_dt"] = pd.to_datetime(
    clean_df["closed_date"], errors="coerce"
)

clean_df = clean_df[
    clean_df["created_dt"].notna()
    & clean_df["created_dt"].ge(pd.Timestamp("2025-01-01"))
    & clean_df["created_dt"].lt(pd.Timestamp("2026-01-01"))
].copy()

record_step(
    2, "Keep valid created_date in 2025",
    before, len(clean_df),
    "กรองวันที่รับแจ้งที่ไม่ถูกต้องและนอกช่วงปีศึกษา"
)

# Step 4: Calculate target and remove missing/invalid close dates
before = len(clean_df)

clean_df["resolution_time"] = (
    clean_df["closed_dt"] - clean_df["created_dt"]
).dt.total_seconds() / 3600

clean_df = clean_df[
    clean_df["closed_dt"].notna()
    & clean_df["resolution_time"].notna()
].copy()

record_step(
    3, "Keep records with valid closed_date",
    before, len(clean_df),
    "ต้องมีวันปิดเรื่องที่ใช้สร้าง Target ได้"
)

# Step 5: Remove negative duration only
before = len(clean_df)

clean_df = clean_df[
    clean_df["resolution_time"] >= 0
].copy()

record_step(
    4, "Remove negative resolution_time",
    before, len(clean_df),
    "วันปิดเรื่องต้องไม่เกิดก่อนวันรับแจ้ง"
)

# Step 6: Fill missing categorical features with Unknown
before = len(clean_df)

feature_cols = ["complaint_type", "agency", "borough"]

for col in feature_cols:
    clean_df[col] = (
        clean_df[col].astype("string").str.strip()
        .replace("", pd.NA)
        .fillna("Unknown")
    )

record_step(
    5, "Fill missing categorical features",
    before, len(clean_df),
    "เก็บรายการไว้และแทนหมวดหมู่ที่ขาดด้วย Unknown"
)

# Step 7: Create time features from created_dt
before = len(clean_df)

clean_df["hour"] = clean_df["created_dt"].dt.hour
clean_df["day_of_week"] = clean_df["created_dt"].dt.dayofweek
clean_df["month"] = clean_df["created_dt"].dt.month

# Northern Hemisphere meteorological seasons
season_map = {
    12: "Winter", 1: "Winter", 2: "Winter",
    3: "Spring", 4: "Spring", 5: "Spring",
    6: "Summer", 7: "Summer", 8: "Summer",
    9: "Fall", 10: "Fall", 11: "Fall"
}

clean_df["season"] = clean_df["month"].map(season_map)

record_step(
    6, "Create date-time features",
    before, len(clean_df),
    "สร้าง hour, day_of_week, month และ season"
)

# Step 8: Report outliers but do not delete automatically
q1 = clean_df["resolution_time"].quantile(0.25)
q3 = clean_df["resolution_time"].quantile(0.75)
iqr = q3 - q1
upper_bound = q3 + 1.5 * iqr

outlier_mask = clean_df["resolution_time"] > upper_bound

print("\n=== Outlier Assessment ===")
print(f"IQR upper bound: {upper_bound:.3f} hours")
print(f"High-duration outliers: {outlier_mask.sum():,}")
print(f"Outlier percentage: {outlier_mask.mean() * 100:.4f}%")
print("หมายเหตุ: รายงาน Outliers แต่ยังไม่ลบออก")

# Final audit table
cleaning_report = pd.DataFrame(cleaning_log)

print("\n=== Data Cleaning Audit Table ===")
display(cleaning_report)

print("\n=== Final Dataset ===")
print(f"Rows: {len(clean_df):,}")
print(f"Columns: {len(clean_df.columns)}")
print(
    "Retention rate:",
    f"{len(clean_df) / initial_rows * 100:.2f}%"
)

# Save cleaned data
clean_df.to_csv("nyc311_2025_cleaned.csv", index=False)

cleaning_report.to_csv(
    "nyc311_2025_cleaning_audit.csv", index=False
)

print("บันทึก nyc311_2025_cleaned.csv เรียบร้อย")
print("บันทึก nyc311_2025_cleaning_audit.csv เรียบร้อย")
```

การตัดตามวันปิดช่วยป้องกันการใช้คำตอบจากอนาคต แต่ไม่ได้แก้ selection bias นี้
ดังนั้นผลจากช่วงฝึกไม่ใช่ข้อสรุปแทนเรื่องที่เปิดค้างหรือทุกเดือนของปี การศึกษาเพียงปีเดียวไม่เพียงพอยืนยันฤดูกาลข้ามปี

```python
# 6.1 ตรวจสัญญาข้อมูลและกำหนดขอบเขต EDA โดยไม่แก้ clean_df ของเพื่อน
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

RANDOM_STATE = 42
PLOT_SAMPLE_SIZE = 50_000
MIN_GROUP_N = 30  # เกณฑ์แสดงกลุ่มให้ไม่ผันผวนมากเกินไป ไม่ใช่การทดสอบนัยสำคัญ
YEAR_START = pd.Timestamp('2025-01-01')
TRAIN_END = pd.Timestamp('2025-09-01')
VAL_END = pd.Timestamp('2025-11-01')
TEST_END = pd.Timestamp('2026-01-01')

if 'clean_df' not in globals():
    # ใช้ไฟล์ cleaned ที่ข้อ 5 บันทึกไว้ได้ เพื่อไม่ต้องดาวน์โหลด API ใหม่ทุกครั้ง
    candidates = [Path('nyc311_2025_cleaned.csv'),
                  Path('data/nyc311_2025_cleaned.csv'),
                  Path('/content/nyc311_2025_cleaned.csv')]
    found = next((p for p in candidates if p.is_file()), None)
    if found is None:
        raise FileNotFoundError(
            'ต้องรันข้อ 3–5 ให้ได้ clean_df หรือวาง nyc311_2025_cleaned.csv '
            'ในโฟลเดอร์ปัจจุบัน/data ก่อนรันข้อ 6–7; ไม่สร้างข้อมูลสมมติแทนข้อมูลจริง'
        )
    clean_df = pd.read_csv(found, dtype={c: 'string' for c in
        ['unique_key', 'complaint_type', 'agency', 'borough', 'incident_zip', 'status']})
    print('อ่านข้อมูล cleaned จาก:', found)

required_cols = ['created_dt', 'closed_dt', 'resolution_time',
                 'complaint_type', 'agency', 'borough']
missing_cols = sorted(set(required_cols) - set(clean_df.columns))
if missing_cols:
    raise ValueError(f'ขาดคอลัมน์จากข้อ 5: {missing_cols}')
if clean_df.empty or not clean_df.index.is_unique:
    raise ValueError('clean_df ต้องไม่ว่างและ index ต้องไม่ซ้ำ เพื่อจับคู่ Features/Target ให้ถูกแถว')

# แปลงสำเนาเฉพาะคอลัมน์ที่ใช้ ไม่ทำสำเนาข้อมูลทั้งชุดหลายครั้ง
created_all = pd.to_datetime(clean_df['created_dt'], errors='coerce')
closed_all = pd.to_datetime(clean_df['closed_dt'], errors='coerce')
target_all = pd.to_numeric(clean_df['resolution_time'], errors='coerce')
if created_all.dt.tz is not None or closed_all.dt.tz is not None:
    raise ValueError('ข้อ 5 ใช้เวลาที่ไม่ระบุ timezone; ต้องตกลง timezone ให้ตรงกันก่อนเปลี่ยนสัญญานี้')
if created_all.isna().any() or closed_all.isna().any():
    raise ValueError('พบวันเวลาที่แปลงไม่ได้ ต้องย้อนตรวจข้อ 5')
if not created_all.between(YEAR_START, TEST_END, inclusive='left').all():
    raise ValueError('พบ created_dt นอกปี 2025')
if not np.isfinite(target_all.to_numpy(dtype=float)).all() or target_all.lt(0).any():
    raise ValueError('Target ต้องเป็นชั่วโมงที่เป็นจำนวนจำกัดและไม่ติดลบ')
expected_target = (closed_all - created_all).dt.total_seconds() / 3600.0
if not np.allclose(target_all.to_numpy(dtype=float), expected_target.to_numpy(),
                   atol=1e-6, rtol=1e-9):
    raise ValueError('resolution_time ไม่ตรงกับผลต่างวันเวลาในหน่วยชั่วโมง')

eda_cohort_mask = created_all.lt(TRAIN_END)
eda_mask = eda_cohort_mask & closed_all.lt(TRAIN_END)
eda_df = clean_df.loc[eda_mask, required_cols].copy()
eda_df['created_dt'] = created_all.loc[eda_mask]
eda_df['closed_dt'] = closed_all.loc[eda_mask]
eda_df['resolution_time'] = target_all.loc[eda_mask].astype(float)
for col in ['complaint_type', 'agency', 'borough']:
    eda_df[col] = (eda_df[col].astype('string').str.strip()
                   .replace('', pd.NA).fillna('Unknown'))
if eda_df.empty:
    raise ValueError('ไม่มีเรื่องที่แจ้งและปิดก่อน TRAIN_END; ตรวจช่วงข้อมูลและข้อกำหนดเวลาร่วมกับกลุ่ม')

eda_df['hour'] = eda_df['created_dt'].dt.hour
eda_df['day_of_week'] = eda_df['created_dt'].dt.dayofweek
eda_df['month'] = eda_df['created_dt'].dt.month
eda_df['is_weekend'] = eda_df['day_of_week'].ge(5).astype('int8')
eda_target = eda_df['resolution_time']
plot_df = eda_df.sample(n=min(PLOT_SAMPLE_SIZE, len(eda_df)), random_state=RANDOM_STATE)
print(f'ขอบเขต EDA: แจ้งและปิดก่อน {TRAIN_END.date()}')
print(f'แถวใน EDA = {len(eda_df):,}; ตัวอย่างเฉพาะกราฟ = {len(plot_df):,}')
print('ไม่ใช้ Target ของ validation/test เพื่อเลือก Features; ไม่ได้เปลี่ยน clean_df')
```

### 6.2 Descriptive Statistics และการกระจายของ Target

รายงานค่าเฉลี่ยควบคู่กับมัธยฐาน P90/P95/P99 เพราะกรณีที่ปิดช้ามากอาจดึงค่าเฉลี่ยขึ้น
ตรวจสัดส่วนศูนย์และค่าที่เกินเกณฑ์ IQR เพื่อเห็นขนาดของปัญหา โดยใช้เกณฑ์นี้เป็นเครื่องมือสำรวจเท่านั้น
หากค่าเฉลี่ยสูงกว่ามัธยฐานมาก ควรเปรียบเทียบ MAE กับ RMSE ในข้อ 10–11 และตรวจข้อผิดพลาดของกรณีปิดช้า
MAE อธิบายความคลาดเคลื่อนเป็นชั่วโมงได้ตรง ส่วน RMSE ลงโทษความผิดพลาดขนาดใหญ่มากกว่า
กราฟ `log1p` แสดงค่าศูนย์ได้และช่วยเห็นโครงสร้างที่ถูกปลายหางบดบัง แต่ไม่ได้เปลี่ยน Target หลักหรือหน่วยประเมิน
หากจะทดลองโมเดลบน `log1p(y)` ให้ทำในข้อ 10 ด้วยตัวแปลง Target และประเมินหลังแปลงกลับเป็นชั่วโมง

```python
target_summary = eda_target.describe(percentiles=[.25, .5, .75, .90, .95, .99]).to_frame('hours')
display(target_summary.round(3))
q1_eda, q3_eda = eda_target.quantile([.25, .75])
iqr_upper_eda = q3_eda + 1.5 * (q3_eda - q1_eda)
zero_n = int(eda_target.eq(0).sum())
high_n = int(eda_target.gt(iqr_upper_eda).sum())
target_diagnostics = pd.DataFrame({
    'measure': ['records', 'zero_duration', 'above_IQR_upper'],
    'count': [len(eda_df), zero_n, high_n],
    'percent': [100.0, 100 * zero_n / len(eda_df), 100 * high_n / len(eda_df)]
})
display(target_diagnostics.round(3))
print(f'IQR upper (EDA เท่านั้น) = {iqr_upper_eda:,.3f} ชั่วโมง; ไม่ลบ/clip ค่าเหล่านี้')
print(f'Skewness (EDA) = {eda_target.skew():.3f}')

# กราฟแรกใช้แกน X แบบ symlog เพื่อแสดงทั้งศูนย์และหางยาว ไม่ตัดทิ้งที่ P99
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
sample_y = plot_df['resolution_time'].to_numpy()
max_y = float(sample_y.max())
edges = np.unique(np.r_[0, np.geomspace(0.001, max(max_y, 0.002), 70)])
axes[0].hist(sample_y, bins=edges, color='#2878A0', edgecolor='white')
axes[0].set_xscale('symlog', linthresh=1)
axes[0].set(xlabel='Resolution time (hours; symlog scale)', ylabel='Sample count',
            title=f'Target distribution (sample n={len(plot_df):,})')
axes[1].hist(np.log1p(sample_y), bins=60, color='#4C956C', edgecolor='white')
axes[1].set(xlabel='log(1 + resolution hours)', ylabel='Sample count',
            title='Log view only: zero durations retained')
fig.tight_layout()
plt.show()
```

### 6.3 โครงสร้าง Features และความแตกต่างระหว่างกลุ่ม

ตรวจจำนวนค่าที่แตกต่างกันก่อน Encoding เพื่อประเมินจำนวนคอลัมน์ที่อาจเพิ่มขึ้นจาก One-hot
ค่าหายที่ข้อ 5 เติม `Unknown` แล้วไม่ใช่ข้อมูลที่กลับมาสมบูรณ์ จึงแสดงจำนวน Unknown แยกด้วย
เลือกแสดง **12 กลุ่มที่มีรายการมากที่สุด** ไม่เลือกเฉพาะกลุ่มที่ Target สูงที่สุด
รายงาน `n`, mean, median และ P90 เพื่อแยก “กลุ่มพบบ่อย” จาก “กลุ่มปิดช้า”
กลุ่ม n < 30 ยังอยู่ในข้อมูล แต่ไม่แสดงในกราฟเปรียบเทียบที่อาจผันผวนสูง
เกณฑ์ n นี้ไม่รับรองความแม่นยำเชิงสถิติ และความต่างระหว่างกลุ่มไม่พิสูจน์ความเป็นเหตุเป็นผล

```python
categorical_eda_cols = ['complaint_type', 'agency', 'borough']
feature_profile = pd.DataFrame([
    {'feature': col, 'dtype': str(eda_df[col].dtype),
     'distinct_values': eda_df[col].nunique(dropna=False),
     'unknown_n': int(eda_df[col].str.casefold().isin(['unknown', 'unspecified']).sum()),
     'unknown_percent': 100 * eda_df[col].str.casefold().isin(['unknown', 'unspecified']).mean()}
    for col in categorical_eda_cols
])
display(feature_profile.round(3))

def summarize_groups(frame, group_col):
    # ไม่ใช้ mean-encoding เป็น Feature; ตารางนี้ใช้เพื่ออธิบาย EDA เท่านั้น
    result = (frame.groupby(group_col, observed=True, dropna=False)['resolution_time']
              .agg(n='size', mean_hours='mean', median_hours='median',
                   p90_hours=lambda s: s.quantile(.90)))
    result['share_percent'] = 100 * result['n'] / len(frame)
    return result.sort_values('n', ascending=False)

group_summaries = {}
for col in categorical_eda_cols:
    summary = summarize_groups(eda_df, col)
    group_summaries[col] = summary  # เก็บครบทุกกลุ่ม ให้ตรวจต่อได้โดยไม่ต้องสุ่ม
    print(f'\n{col}: ตาราง 15 กลุ่มที่พบบ่อย; ตารางเต็มอยู่ใน group_summaries')
    display(summary.head(15).round(3))
    shown = summary.loc[summary['n'].ge(MIN_GROUP_N)].head(12).iloc[::-1]
    if shown.empty:
        print(f'ข้ามกราฟ {col}: ไม่มีกลุ่มที่ n >= {MIN_GROUP_N}')
        continue
    fig, axes = plt.subplots(1, 2, figsize=(14, max(3.5, len(shown) * .4)))
    positions = np.arange(len(shown))
    labels = [str(v) for v in shown.index]
    axes[0].barh(positions, shown['n'], color='#2878A0')
    axes[0].set_yticks(positions)
    axes[0].set_yticklabels(labels)
    axes[0].set(xlabel='Count (all EDA records)', title=f'{col}: most frequent groups')
    axes[1].scatter(shown['median_hours'], positions, color='#4C956C', label='Median', zorder=3, s=40)
    axes[1].scatter(shown['p90_hours'], positions, color='#C65D3B', label='P90', zorder=3)
    axes[1].set_yticks(positions)
    axes[1].set_yticklabels([f'{l[:26]} (n={v:,})' for l, v in zip(labels, shown['n'])])
    axes[1].set_xscale('log')
    axes[1].set(xlabel='Resolution hours (log scale)', title='Typical duration (median) and upper tail (P90)')
    axes[1].legend()
    fig.tight_layout()
    plt.show()
```

### 6.4 ความสัมพันธ์ระหว่างเวลารับแจ้งกับ Target

ใช้ชั่วโมง วันในสัปดาห์ และเดือนจาก **เวลารับแจ้ง** ไม่ใช้เวลาปิดเรื่อง
ทุกกราฟแสดงจำนวนรายการคู่กับ median/P90 เพื่อไม่ตีความเดือนหรือช่วงเวลาที่มีข้อมูลน้อยเกินไป
เดือนในช่วง EDA นี้มีเพียง ม.ค.–ส.ค. จึงห้ามอ้างว่าพิสูจน์พฤติกรรมตลอดปีหรือทุกฤดูกาลแล้ว
**ข้อควรระวังในการอ่านกราฟเดือน:** ค่ามัธยฐาน/P90 ที่ลดลงจาก ม.ค. ถึง ส.ค. ส่วนหนึ่งเป็นผลของการตัดทอน (right-censoring) เพราะ EDA นี้นับเฉพาะเรื่องที่ปิดก่อน 1 ก.ย. เรื่องที่แจ้งใกล้เดือน ส.ค. จึงมีเวลาปิดได้น้อยกว่า (ดูสัดส่วน coverage ในข้อ 6.5) และกราฟชั่วโมง/วันในสัปดาห์ปนกับสัดส่วนประเภทเรื่อง (เช่น เรื่องของ NYPD ที่ปิดเร็วมักแจ้งช่วงกลางคืน/สุดสัปดาห์) จึงไม่ควรตีความเป็นฤดูกาลหรือผลของเวลาโดยตรง ข้อ 10 จะทดสอบด้วย ablation แทน

ไม่ใช้ Pearson correlation กับรหัสหมวดหมู่ที่กำหนดตัวเลขเอง เพราะลำดับตัวเลขไม่ได้แปลว่ามีระยะห่างจริง

```python
time_specs = [('hour', range(24)), ('day_of_week', range(7)), ('month', range(1, 9))]
time_summaries = {}
fig, axes = plt.subplots(3, 2, figsize=(14, 11))
for row, (col, expected_values) in enumerate(time_specs):
    result = summarize_groups(eda_df, col).reindex(expected_values)
    result['n'] = result['n'].fillna(0).astype(int)
    time_summaries[col] = result
    display(result.round(3))
    axes[row, 0].bar(result.index, result['n'], color='#2878A0')
    axes[row, 0].set(xlabel=col, ylabel='Count', title=f'Intake timing: {col}')
    reliable = result['n'].ge(MIN_GROUP_N)
    axes[row, 1].plot(result.index, result['median_hours'].where(reliable),
                       marker='o', label='Median', color='#4C956C')
    axes[row, 1].plot(result.index, result['p90_hours'].where(reliable),
                       marker='s', label='P90', color='#C65D3B')
    axes[row, 1].set(xlabel=col, ylabel='Resolution hours',
                     title=f'Duration by {col} (groups n >= {MIN_GROUP_N}; log y)')
    axes[row, 1].set_yscale('log')
    for ax in axes[row]:
        ax.set_xticks(list(expected_values))
    axes[row, 1].legend()
axes[1, 0].set_xticklabels(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
axes[1, 1].set_xticklabels(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
fig.tight_layout()
plt.show()

print('เปรียบเทียบวันธรรมดา/สุดสัปดาห์: 0 = Mon–Fri, 1 = Sat–Sun')
weekend_summary = summarize_groups(eda_df, 'is_weekend')
display(weekend_summary.round(3))
```

### 6.5 ตรวจผลของการมีคำตอบทันวันฝึก และสรุป EDA จากค่าที่คำนวณจริง

ตรวจสัดส่วนเรื่องที่ปิดทันวันฝึกแยกตามเดือนรับแจ้ง เพราะกลุ่มที่แจ้งใกล้วันแบ่งมีเวลารอปิดสั้นกว่า
หาก `raw_df` ยังอยู่ จะใช้เป็นตัวหารเพื่อรวมเรื่องที่ยังไม่ปิดด้วย โดยอ่านเฉพาะวันรับแจ้งและวันปิด
หากมีเพียง `clean_df` จะรายงานสัดส่วนภายในกลุ่มที่ผ่านข้อ 5 พร้อมระบุว่าตัวหารไม่ครอบคลุมเรื่องเปิดค้าง

สรุปด้านล่างสร้างจากผลรันโดยตรง ไม่คัดลอกจำนวนของข้อมูลทั้งปีมาอ้างเป็นผลของช่วงฝึก

```python
if 'raw_df' in globals() and {'created_date', 'closed_date'}.issubset(raw_df.columns):
    audit_created = pd.to_datetime(raw_df['created_date'], errors='coerce')
    audit_closed = pd.to_datetime(raw_df['closed_date'], errors='coerce')
    audit_scope = audit_created.ge(YEAR_START) & audit_created.lt(TRAIN_END)
    coverage_source = 'raw_df: รวมเรื่องที่ยังไม่มีวันปิด (ก่อนตรวจแถวซ้ำตามข้อ 5)'
else:
    audit_created, audit_closed = created_all, closed_all
    audit_scope = eda_cohort_mask
    coverage_source = 'clean_df เท่านั้น: ไม่รวมเรื่องที่ข้อ 5 ตัดออกเพราะไม่มีวันปิด/วันไม่ถูกต้อง'

coverage = pd.DataFrame({
    'month': audit_created.loc[audit_scope].dt.month,
    'known_by_train_end': (
        audit_closed.loc[audit_scope].notna()
        & audit_closed.loc[audit_scope].ge(audit_created.loc[audit_scope])
        & audit_closed.loc[audit_scope].lt(TRAIN_END)
    ).astype('int8')
})
closure_coverage = coverage.groupby('month', observed=True)['known_by_train_end'].agg(
    intake_n='size', label_known_n='sum')
closure_coverage['known_percent'] = 100 * closure_coverage['label_known_n'] / closure_coverage['intake_n']
print('ตัวหารของตาราง coverage:', coverage_source)
display(closure_coverage.round(3))
del coverage, audit_created, audit_closed, audit_scope


grp = group_summaries['complaint_type']
big = grp.loc[grp['n'] >= 1000]
agy = group_summaries['agency']
agy_big = agy.loc[agy['n'] >= 1000]
wk = weekend_summary['median_hours']
mon = time_summaries['month']
cov_min = closure_coverage['known_percent'].min()
cov_min_month = int(closure_coverage['known_percent'].idxmin())

eda_findings = [
    (f"Target เบ้ขวามาก: skew {eda_target.skew():.2f}, ค่าเฉลี่ย {eda_target.mean():,.1f} ชม. เทียบมัธยฐาน {eda_target.median():,.2f} ชม., P99 {eda_target.quantile(.99):,.0f} ชม.",
     "ฝึกบน log1p(y) แล้วแปลงกลับเป็นชั่วโมงก่อนวัดผล (ข้อ 9) รายงาน MAE, MedAE, RMSE ควบคู่กัน และเทียบกับ Dummy median (ข้อ 10–11)"),
    (f"ประเภทเรื่องและหน่วยงานต่างกันมาก: ใน {len(big)} ประเภทที่ n≥1,000 มัธยฐานตั้งแต่ {big['median_hours'].min():.2f} ถึง {big['median_hours'].max():,.1f} ชม.; หน่วยงาน {agy_big['median_hours'].min():.2f} ถึง {agy_big['median_hours'].max():,.1f} ชม. ฮิสโตแกรม Target มีหลายยอดสอดคล้องกัน",
     "ใช้ complaint_type, agency, borough เป็น Features หลัก (One-hot แบบ sparse, min_frequency) และถือชุด base เป็นจุดตั้งต้นของ ablation"),
    (f"ช่วงเวลาแจ้ง: มัธยฐานวันธรรมดา {wk.get(0, float('nan')):.2f} ชม. เทียบสุดสัปดาห์ {wk.get(1, float('nan')):.2f} ชม. แต่รูปแบบนี้ปนกับสัดส่วนประเภทเรื่อง",
     "สร้างฟีเจอร์เวลา (sin/cos, is_weekend, is_business_hour) แล้ววัดด้วย ablation ว่ายังช่วยหลังควบคุมประเภทเรื่องหรือไม่ ไม่สรุปจาก EDA อย่างเดียว"),
    (f"แนวโน้มรายเดือน: มัธยฐานเดือนแรก {mon['median_hours'].iloc[0]:.2f} ชม. → เดือนสุดท้าย {mon['median_hours'].iloc[-1]:.2f} ชม. แต่สัดส่วนเรื่องที่มีคำตอบทันวันตัดต่ำสุด {cov_min:.1f}% (เดือน {cov_min_month}) จึงเป็นผลของ right-censoring",
     "ไม่ตีความเดือนเป็นฤดูกาล, กำหนดกฎ train/validation/test ตามเวลา (ข้อ 8), ทดสอบ ablation ถอดฟีเจอร์เดือน และรายงานว่าข้อมูลปีเดียวยืนยัน seasonality ไม่ได้"),
    (f"ระยะเวลาศูนย์ {zero_n:,} เรื่อง และสูงกว่า IQR upper {high_n:,} เรื่อง ({100 * high_n / len(eda_df):.1f}%)",
     "เก็บไว้ทั้งหมด ไม่ clip; ใช้ MAE/MedAE เป็นหลักและวิเคราะห์ error แยกกลุ่มปิดช้า (ข้อ 11.4)"),
]
_md = ["**สิ่งที่ค้นพบจาก EDA และผลต่อการเตรียมข้อมูล/การเลือกโมเดล**", "",
       "| # | สิ่งที่ค้นพบ | ผลต่อการทำงานต่อ |", "|---:|---|---|"]
for i, (a, b) in enumerate(eda_findings, 1):
    _md.append(f"| {i} | {a} | {b} |")
display(Markdown("\n".join(_md)))
```

## 7. Feature Engineering

### 7.1 หลักการและรายการ Features พร้อมเหตุผล

สร้างจากข้อมูลที่คาดว่าจะมี ณ **เวลารับแจ้ง** เท่านั้น โดยฟังก์ชันไม่อ่าน `closed_dt`, `closed_date`,
`status`, `unique_key` หรือ Target การสร้างปฏิทิน/ไซน์/โคไซน์เป็นกฎคงที่ ไม่ต้อง fit จากข้อมูล
จึงสร้างก่อนแบ่งชุดได้ ส่วนการเติมค่าจากสถิติ, Encoding, Scaling, การรวมหมวดหมู่หายาก,
Feature selection และ Target transformation ที่ต้องเรียนรู้ ให้ทำในข้อ 9–10 ภายในแต่ละ training fold

| Feature / กลุ่ม | สร้างจาก | เหตุผล | วิธีใช้ต่อ |
|---|---|---|---|
| `complaint_type` | ประเภทเรื่องรับแจ้ง | เนื้องานและขั้นตอนดำเนินการต่างกัน; EDA ใช้ตรวจความแตกต่างพร้อม n | One-hot ภายใน Pipeline; ไม่แทนด้วยเลขลำดับหมวดหมู่ |
| `agency` | หน่วยงานที่รับเรื่อง | อาจสะท้อนกระบวนการและภาระงานต่างกัน แต่ **ต้องยืนยันว่าเป็นหน่วยงาน ณ รับแจ้ง ไม่ใช่หน่วยงานหลังส่งต่อ** | One-hot; ถ้ายืนยันไม่ได้ ให้ทดลองตัดออกใน validation และรายงานข้อจำกัด |
| `borough` | เขตพื้นที่รับแจ้ง | อาจสะท้อนบริบทพื้นที่และทรัพยากร; ไม่ใช่ข้อพิสูจน์ว่าพื้นที่เป็นสาเหตุ | One-hot; รักษา Unknown/Unspecified และไม่เดาพื้นที่จากรหัส |
| `hour`, `day_of_week`, `month` | `created_dt` | อธิบายเวลาแจ้งในหน่วยที่อ่านได้ และให้ต้นไม้แบ่งช่วง | ใช้ในชุด `tree`; Monday=0, Sunday=6 |
| `season` | กฎเดือนแบบฤดูกาลอุตุนิยมวิทยาซีกโลกเหนือ | สรุปเดือนเป็นฤดูกาลเพื่อ EDA/การแปลผล; เป็นตัวแปรซ้ำเชิงข้อมูลกับ month | เก็บสำหรับทดลองเพิ่มภายหลัง ไม่ใส่ baseline คู่กับ month โดยไม่มีหลักฐาน |
| `is_weekend` | วันในสัปดาห์ >= 5 | สมมติฐานเรื่องรอบงานวันธรรมดา/สุดสัปดาห์ | ตัวบ่งชี้ 0/1; ต้องวัดประโยชน์จาก validation |
| `is_business_hour` | จ.–ศ. และ 09:00–16:59 | ตัวแทนช่วงเวลางานแบบคงที่ | เป็นเพียง proxy ไม่ใช่เวลาทำงานจริงของทุกหน่วยงาน และยังไม่หักวันหยุด |
| `hour_sin`, `hour_cos` | sin/cos(2π × hour / 24) | ทำให้ 23:00 กับ 00:00 อยู่ใกล้กัน แทนการมองว่าห่าง 23 ชั่วโมง | ใช้คู่กันในชุด `cyclic` สำหรับโมเดลเชิงเส้น/ระยะทาง |
| `dow_sin`, `dow_cos` | sin/cos(2π × weekday / 7) | เชื่อม Sunday กับ Monday โดยไม่กำหนดระยะเชิงเส้นผิด | ใช้คู่กันในชุด `cyclic` |
| `month_sin`, `month_cos` | sin/cos(2π × (month−1) / 12) | เชื่อม December กับ January และใช้กฎเดียวกับเดือนอนาคต | ใช้คู่กันในชุด `cyclic`; ข้อมูลปีเดียวไม่รับรอง seasonality |

**ชุดเสริมสำหรับ ablation (ข้อ 10.6):** `cyclic_nomonth` (ไม่มี month_sin/cos) และ `cyclic_season` (cyclic + `season` แบบ One-hot) ใช้ตอบว่าเวลา/เดือน/ฤดูกาลช่วยทำนายจริงหรือไม่

**ชุด Features สำหรับเปรียบเทียบอย่างเป็นธรรม:**
- `base`: ประเภทเรื่อง หน่วยงาน และพื้นที่ เป็นจุดตั้งต้นเพื่อวัดว่าตัวแปรเวลาช่วยจริงหรือไม่
- `tree`: `base` + ชั่วโมง/วัน/เดือน + ตัวบ่งชี้ เหมาะกับ Decision Tree/Random Forest/Boosting ที่แบ่งช่วงได้
- `cyclic`: `base` + sin/cos ทั้ง 3 รอบ + ตัวบ่งชี้ สำหรับ Linear/Ridge หรือโมเดลที่ใช้ระยะทาง

ทั้งสามเป็น **ชุดผู้สมัคร** ไม่ใช่ข้อสรุปว่าใช้ชุดใดดีที่สุด ห้ามเลือกจากคะแนน test
หลีกเลี่ยงใส่ดิบ+cyclic+season ทั้งหมดโดยอัตโนมัติ เพราะเพิ่มความซ้ำซ้อนและทำให้แปลความสำคัญยากขึ้น

```python
# 7.2 Pure feature function: ใช้กับข้อมูลใหม่ได้โดยไม่ต้องมีวันปิดหรือ Target
INTAKE_CATEGORY_COLS = ['complaint_type', 'agency', 'borough']

def make_intake_features(frame):
    needed = ['created_dt'] + INTAKE_CATEGORY_COLS
    missing = sorted(set(needed) - set(frame.columns))
    if missing:
        raise ValueError(f'ข้อมูลรับแจ้งขาดคอลัมน์: {missing}')
    if not frame.index.is_unique:
        raise ValueError('index ซ้ำ ทำให้จับคู่ผลทำนายกับเรื่องร้องเรียนกำกวม')
    dt = pd.to_datetime(frame['created_dt'], errors='coerce')
    if dt.isna().any():
        raise ValueError('created_dt ไม่ถูกต้อง ต้องแก้ก่อนสร้าง Features')
    if dt.dt.tz is not None:
        raise ValueError('ต้องใช้เวลาในระบบเดียวกับข้อ 5; ห้ามเปลี่ยน timezone โดยเงียบ ๆ')

    result = pd.DataFrame(index=frame.index)
    for col in INTAKE_CATEGORY_COLS:
        # ค่าคงที่ Unknown ไม่ได้เรียนรู้จากทั้งชุด; ไม่รวม case ต่างกันแบบเดาสุ่ม
        result[col] = (frame[col].astype('string').str.strip()
                       .replace('', pd.NA).fillna('Unknown').astype(object))
    result['hour'] = dt.dt.hour.astype('int8')
    result['day_of_week'] = dt.dt.dayofweek.astype('int8')
    result['month'] = dt.dt.month.astype('int8')
    result['season'] = result['month'].map({
        12: 'Winter', 1: 'Winter', 2: 'Winter',
        3: 'Spring', 4: 'Spring', 5: 'Spring',
        6: 'Summer', 7: 'Summer', 8: 'Summer',
        9: 'Fall', 10: 'Fall', 11: 'Fall'
    }).astype(object)
    result['is_weekend'] = result['day_of_week'].ge(5).astype('int8')
    result['is_business_hour'] = (
        result['day_of_week'].lt(5) & result['hour'].between(9, 16)
    ).astype('int8')
    for source_col, prefix, period, offset in [
        ('hour', 'hour', 24, 0), ('day_of_week', 'dow', 7, 0), ('month', 'month', 12, 1)
    ]:
        angle = 2 * np.pi * (result[source_col].astype(float) - offset) / period
        result[f'{prefix}_sin'] = np.sin(angle).astype('float32')
        result[f'{prefix}_cos'] = np.cos(angle).astype('float32')
    return result

feature_df = make_intake_features(clean_df)
feature_sets = {
    'base': INTAKE_CATEGORY_COLS.copy(),
    'tree': INTAKE_CATEGORY_COLS + ['hour', 'day_of_week', 'month',
                                   'is_weekend', 'is_business_hour'],
    'cyclic': INTAKE_CATEGORY_COLS + ['hour_sin', 'hour_cos', 'dow_sin', 'dow_cos',
                                     'month_sin', 'month_cos', 'is_weekend', 'is_business_hour']
}
feature_sets['cyclic_nomonth'] = INTAKE_CATEGORY_COLS + ['hour_sin', 'hour_cos', 'dow_sin', 'dow_cos',
                                                         'is_weekend', 'is_business_hour']
feature_sets['cyclic_season'] = feature_sets['cyclic'] + ['season']

# X เป็นตารางผู้สมัคร; ข้อ 8–10 ต้องเลือก feature_sets[...] แทนใช้ทุกคอลัมน์อัตโนมัติ
X = feature_df
y = target_all.rename('resolution_time').copy()

# แยก metadata ไว้กำหนดเวลา/ติดตามแถว ห้ามส่งให้โมเดล
split_info = pd.DataFrame({'created_dt': created_all, 'closed_dt': closed_all}, index=clean_df.index)
if 'unique_key' in clean_df.columns:
    split_info['unique_key'] = clean_df['unique_key'].astype('string')

feature_groups = {
    name: {
        'categorical': [c for c in cols if c in INTAKE_CATEGORY_COLS or c == 'season'],
        'numeric': [c for c in cols if c not in INTAKE_CATEGORY_COLS and c != 'season']
    }
    for name, cols in feature_sets.items()
}
print('Feature candidates:', feature_df.shape)
print('Target:', y.shape, '(ชั่วโมง)')
for name, cols in feature_sets.items():
    print(f'{name}: {len(cols)} Features = {cols}')
display(feature_df.head())
```

### 7.4 ส่งต่องานให้ข้อ 8–10

| ตัวแปรที่เตรียมไว้ | ความหมาย/วิธีใช้ |
|---|---|
| `clean_df` | ข้อมูลหลังทำความสะอาดของเพื่อน ไม่แก้ไขจากข้อ 6–7 |
| `feature_df` หรือ `X` | ตาราง Features ผู้สมัคร **เลือกคอลัมน์จาก feature_sets ก่อนฝึก** |
| `y` | `resolution_time` หน่วยชั่วโมง ตาม index ของ X |
| `feature_sets`, `feature_groups` | รายชื่อชุดผู้สมัครและตัวแยก categorical/numeric สำหรับ ColumnTransformer |
| `split_info` | เวลาแจ้ง เวลาเสร็จ และรหัสรายการ ใช้แบ่งชุด/ตรวจข้ามชุดเท่านั้น ห้ามนำไปเข้าโมเดล |
| `TRAIN_END`, `VAL_END`, `TEST_END` | 2025-09-01, 2025-11-01, 2026-01-01 ตามลำดับ |
| `group_summaries`, `time_summaries`, `closure_coverage`, `eda_findings` | หลักฐาน EDA ช่วงฝึก พร้อมใช้แปลผล/นำเสนอเมื่อรันจริง |

**สัญญาที่ข้อ 8 ต้องรักษา** (แสดงแนวทาง ไม่ได้เขียนแทนส่วนงานของเพื่อน):
1. Train: รับแจ้ง 1 ม.ค.–ก่อน 1 ก.ย. และ `closed_dt < TRAIN_END` จึงมีคำตอบให้เรียน ณ วันฝึก
2. Validation: รับแจ้ง 1 ก.ย.–ก่อน 1 พ.ย. และ `closed_dt < VAL_END` จึงมีคำตอบให้จูน ณ วันเลือกโมเดล
3. Test: รับแจ้ง 1 พ.ย.–ก่อน 1 ม.ค. 2026 กันไว้ประเมินสุดท้ายเท่านั้น ต้องระบุวัน snapshot/วันสิ้นสุดการติดตามผลจริง
   ห้ามสมมติว่าเรื่องที่ยังไม่ปิดใช้เวลาเป็นศูนย์ และห้ามตัด Target ยาวเพียงเพื่อให้คะแนนดีขึ้น
4. กรณีจูนด้วย cross-validation ให้เรียงตาม `created_dt` และใช้ rolling/forward validation
   พร้อมคัดแถวฝึกที่วันปิดยังไม่ถึงต้น validation fold ออก (label availability)
   TimeSeriesSplit ที่เรียงตามวันแจ้งอย่างเดียว **ยังไม่พอ** หากนำคำตอบที่ปิดภายหลังมาใช้ฝึก
5. ถ้ารวม train+validation เพื่อ refit ก่อนทดสอบ ต้องเลือกเฉพาะเรื่องที่แจ้งก่อน `VAL_END` และปิดก่อน `VAL_END`
   ไม่ใช้คำตอบจากหลังวันเริ่ม test ย้อนกลับไปฝึก
6. เลือก Features/Encoding/Scaling จาก train เท่านั้น และ fit ใหม่ภายในแต่ละ fold; validation/test ใช้ transform
   สำหรับ One-hot ให้รองรับหมวดหมู่ใหม่ด้วย `handle_unknown='ignore'` และบันทึกอัตราหมวดหมู่ที่ไม่เคยพบ
   หากรวมหมวดหมู่หายาก ให้หาเกณฑ์จาก train เท่านั้น ไม่ใช้ความถี่ทั้งชุด

**แนวทางการเลือกโมเดลตามบทเรียน:** เปรียบเทียบ DummyRegressor (median สำหรับ baseline ของ MAE)
กับอย่างน้อย 3 โมเดลตามข้อ 10 เช่น Ridge, Decision Tree และ Random Forest/Gradient Boosting
ใช้ชุด cyclic กับ Ridge และชุด tree กับต้นไม้ ทดลองเพิ่มเวลาเทียบ base ด้วย validation เดียวกัน
อย่าสรุปว่าโมเดลซับซ้อนดีกว่าจนกว่าจะมีตารางคะแนน/เวลา/ความคลาดเคลื่อนรายกลุ่มจริง
คำนึงถึงข้อมูลหลักล้านแถว: ใช้ One-hot แบบ sparse ที่เข้ากับโมเดล หลีกเลี่ยงแปลงทั้งชุดเป็น dense โดยไม่ประเมิน RAM
การสุ่มเพื่อทดลองโมเดลต้องทำจาก train เท่านั้นและรายงานจำนวนที่ใช้; ตัวอย่างสุ่มในข้อ 6 ไม่ใช่ training set ของโมเดล

**ข้อจำกัดที่คงอยู่:** ไม่มีวัน snapshot และประวัติการเปลี่ยนหน่วยงานในตาราง 8 คอลัมน์
จึงยังพิสูจน์ความครบของ Target/ค่าของ Features ณ รับแจ้งไม่ได้ ต้องระบุในรายงาน
การเลือก cohort ที่ปิดแล้วอาจทำให้กรณีที่ปิดช้ามากถูกแทนไม่ครบ หากต้องตอบเรื่องที่ยังเปิดค้างอย่างจริงจัง
ควรศึกษา survival analysis หรือออกแบบช่วงติดตามผลที่เท่ากันเป็นงานต่อยอด ไม่อ้างว่า Regression ปัจจุบันแก้แล้ว

**วิธีแบ่ง: แบ่งตามลำดับเวลา (Chronological Split) ตามวันที่รับแจ้ง (`created_dt`)**

| ชุดข้อมูล | ช่วงวันที่รับแจ้ง | หน้าที่ |
|---|---|---|
| Train | 1 ม.ค. – 31 ส.ค. 2025 | ฝึกโมเดล |

**เหตุผลที่วิธีนี้เหมาะกับข้อมูลและคำถามของกลุ่ม**

1. **คำถามคือการทำนายเรื่องใหม่ในอนาคต** ชุดทดสอบจึงต้องเป็นช่วงเวลาหลังชุดฝึก การสุ่มแบ่งจะทำให้เรื่องช่วงเดียวกันปนอยู่ทั้งในชุดฝึกและทดสอบ (temporal leakage) ได้คะแนนดีเกินจริง
2. **ข้อมูลมีขนาดใหญ่** ทุกชุดมีหลักแสนถึงหลักล้านแถว เพียงพอสำหรับฝึก เลือกโมเดล และประเมินผล
3. **การจัดการ label ที่ปิดหลังวันตัดแบ่ง (การตัดสินใจที่ตั้งใจ):** แบ่งตามวันรับแจ้งอย่างเดียว ไม่ตัดเรื่องที่แจ้งก่อนวันแบ่งแต่ปิดหลังวันแบ่งออก เพราะการตัดจะทำให้ชุดฝึกขาดเรื่องที่ปิดช้า (ซึ่งเป็นกรณีที่ทำนายยากที่สุด) ชุดฝึกจึงใช้ค่า Target ตามที่ทราบ ณ วัน snapshot ของข้อมูล **ข้อควรระวัง:** ในการใช้งานจริง ณ วันแบ่ง จะยังไม่รู้ Target ของเรื่องเหล่านี้ กลุ่มจึงทำ **sensitivity check ในข้อ 11.5** ที่ฝึกใหม่ด้วยกฎเข้ม (ใช้เฉพาะเรื่องที่ปิดก่อนวันแบ่ง) เพื่อวัดว่าการตัดสินใจนี้เปลี่ยนผลสรุปหรือไม่

## 9. Preprocessing Pipeline

ขั้นตอนที่ต้อง "เรียนรู้จากข้อมูล" ได้แก่ การเติมค่าว่าง การเข้ารหัสหมวดหมู่ และการปรับสเกล ทั้งหมดทำ `.fit()` เฉพาะบน **Train Set** แล้วนำไป `.transform()` กับ Validation และ Test ด้วย `Pipeline` / `ColumnTransformer` ของ scikit-learn

| ขั้นตอน | ตัวแปร | วิธี | สิ่งที่เรียนรู้จาก Train |
|---|---|---|---|
| Missing values | `complaint_type`, `agency`, `borough` | `SimpleImputer(strategy='constant', fill_value='Unknown')` | ไม่ต้องเรียนรู้ (ค่าคงที่) |
| Missing values | ตัวแปรเวลาเชิงตัวเลข | `SimpleImputer(strategy='median')` | ค่ามัธยฐานของแต่ละตัวแปร |
| Encoding | ตัวแปรหมวดหมู่ | `OneHotEncoder(handle_unknown='ignore', min_frequency=100)` แบบ Sparse | รายการหมวดหมู่ และหมวดที่พบน้อยกว่า 100 แถว |
| Scaling | ตัวแปร sin/cos (ชุด `cyclic` สำหรับ Ridge) | `StandardScaler` | ค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐาน |
| Target transformation | `resolution_time` | `log1p` ผ่าน `TransformedTargetRegressor` | ไม่ต้องเรียนรู้ (ฟังก์ชันคงที่) |

**ข้อสังเกตและการตัดสินใจ**
- ชุด `tree` ไม่ต้อง Scale เพราะต้นไม้ไม่ไวต่อสเกลของตัวแปร
- Target เบ้ขวามาก (ค่าเฉลี่ยของ Train สูงกว่ามัธยฐานหลายสิบเท่า ดูตารางข้อ 8 และ skew ในข้อ 6.2) จึงฝึกบน `log1p(y)` แล้วแปลงกลับเป็นชั่วโมงก่อนคำนวณ Metric
- ไม่ใช้ SMOTE เพราะ Target เป็นค่าต่อเนื่อง ไม่ใช่ปัญหา Class Imbalance

```python
# ข้อ 9: Preprocessing Pipeline (fit เฉพาะ train เท่านั้น แล้ว transform val/test)
import scipy.sparse as sp
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# ---------- 1) นิยาม Pipeline ----------
def make_preprocessor(set_name, scale=False):
    """One-hot (sparse) สำหรับ Ridge / Decision Tree"""
    g = feature_groups[set_name]
    cat = Pipeline([
        ('imp', SimpleImputer(strategy='constant', fill_value='Unknown')),
        ('ohe', OneHotEncoder(handle_unknown='ignore', min_frequency=100)),
    ])
    transformers = [('cat', cat, g['categorical'])]
    if g['numeric']:
        steps = [('imp', SimpleImputer(strategy='median'))]
        if scale:
            steps.append(('scaler', StandardScaler()))
        transformers.append(('num', Pipeline(steps), g['numeric']))
    return ColumnTransformer(transformers)

def make_ordinal_preprocessor(set_name):
    """Ordinal สำหรับ HistGradientBoosting (รองรับ categorical โดยตรง)"""
    g = feature_groups[set_name]
    cat = Pipeline([
        ('imp', SimpleImputer(strategy='constant', fill_value='Unknown')),
        ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1,
                               min_frequency=200)),
    ])
    transformers = [('cat', cat, g['categorical'])]
    if g['numeric']:
        transformers.append(('num', SimpleImputer(strategy='median'), g['numeric']))
    return ColumnTransformer(transformers)

def with_log_target(pipe):
    """ฝึกบน log1p(y) แล้วแปลงกลับเป็นชั่วโมงด้วย expm1 ตอนทำนาย"""
    return TransformedTargetRegressor(regressor=pipe, func=np.log1p, inverse_func=np.expm1)

# ---------- 2) อัตราหมวดหมู่ที่ไม่เคยพบใน train ----------
unseen_rows = []
for col in ['complaint_type', 'agency', 'borough']:
    seen = set(X_train[col])
    unseen_rows.append({
        'column': col,
        'train_categories': len(seen),
        'validation_unseen_%': 100 * (~X_val[col].isin(seen)).mean(),
        'test_unseen_%': 100 * (~X_test[col].isin(seen)).mean(),
    })
display(pd.DataFrame(unseen_rows).round(4))
print('Target เป็นค่าต่อเนื่อง จึงไม่ใช้ SMOTE / ไม่ปรับสมดุลคลาส\n')

# ---------- 3) สาธิต: fit บน Train เท่านั้น แล้ว transform Validation/Test ----------
pp_shapes = {}
unseen_df = pd.DataFrame(unseen_rows)
for set_name, scale in [('cyclic', True), ('tree', False)]:
    cols = feature_sets[set_name]
    pre = make_preprocessor(set_name, scale=scale)

    Xtr = pre.fit_transform(X_train[cols])   # fit เรียนรู้จาก Train เท่านั้น
    Xva = pre.transform(X_val[cols])         # Validation: transform อย่างเดียว
    Xte = pre.transform(X_test[cols])        # Test: transform อย่างเดียว

    ohe = pre.named_transformers_['cat'].named_steps['ohe']
    pp_shapes[set_name] = (Xtr.shape[1], ohe.get_feature_names_out().shape[0])
    print(f'--- ชุด {set_name} (scale={scale}) ---')
    print(f'Train      : {Xtr.shape}, sparse={sp.issparse(Xtr)}')
    print(f'Validation : {Xva.shape}')
    print(f'Test       : {Xte.shape}')
    print(f'คอลัมน์ One-hot ที่เรียนรู้จาก Train: {ohe.get_feature_names_out().shape[0]} คอลัมน์')
    for name, M in [('train', Xtr), ('validation', Xva), ('test', Xte)]:
        data = M.data if sp.issparse(M) else np.asarray(M)
        assert np.isfinite(data).all(), f'{name} มี NaN/inf หลัง transform'
    print('ตรวจแล้ว: ไม่มี NaN/inf ในทั้งสามชุด')

    if scale:
        sc = pre.named_transformers_['num'].named_steps['scaler']
        num_cols = feature_groups[set_name]['numeric']
        check = pd.DataFrame({
            'mean_ที่_scaler_เรียนรู้': sc.mean_,
            'mean_จริงของ_train': X_train[num_cols].mean().values,
            'mean_ของ_validation': X_val[num_cols].mean().values,
        }, index=num_cols).round(4)
        display(check)
        print('mean ที่ scaler ใช้ตรงกับ Train ไม่ใช่ Validation/Test')
```

## 10. Model Selection and Hyperparameter Tuning

### 10.1 โมเดลที่เลือกและเหตุผล

| โมเดล | เหตุผลที่เลือก |
|---|---|
| DummyRegressor (median) | Baseline ไว้พิสูจน์ว่าโมเดลที่สร้างเหนือกว่าการเดาแบบง่าย |
| Ridge Regression | โมเดลเชิงเส้นพื้นฐาน L2 Penalty ช่วยเมื่อมี One-hot หลายร้อยคอลัมน์และตัวแปร sin/cos ที่สัมพันธ์กัน (ใช้ชุด `cyclic`) |
| Decision Tree | จับความสัมพันธ์ไม่เชิงเส้นและปฏิสัมพันธ์ระหว่างประเภทเรื่อง หน่วยงาน พื้นที่ และเวลา อธิบายผลได้ง่าย (ใช้ชุด `tree`) |
| Random Forest | เฉลี่ยหลายต้นไม้เพื่อลดความแปรปรวนของต้นไม้เดี่ยว เหมาะกับข้อมูลตารางที่มีตัวแปรหมวดหมู่ (ใช้ชุด `tree`) |

### 10.2 วิธีปรับจูน (ไม่ใช้ Test Set)

ปรับจูนด้วย **Validation Set** เท่านั้น โดยฝึกบนตัวอย่างสุ่มจาก Train Set จำนวน 400,000 แถว แล้ววัดผลบน Validation Set เต็ม ไม่แตะ Test Set ทุกโมเดลฝึกบน `log1p(resolution_time)` แล้วแปลงกลับเป็นชั่วโมงก่อนคำนวณ Metric

| โมเดล | พารามิเตอร์ที่ทดลอง |
|---|---|
| Ridge | alpha = 1, 10, 100 |
| Decision Tree | max_depth = 8, 12, 16; min_samples_leaf = 100, 500 |
| Decision Tree (ชุด base) | max_depth = 10; min_samples_leaf = 200 (ใช้เทียบว่าตัวแปรเวลาช่วยหรือไม่) |
| Random Forest | max_depth = 12, 16; min_samples_leaf = 100; n_estimators = 100 |

**เพิ่มเติมหลังการทดลองหลัก:** ข้อ 10.5 ขยายช่วง alpha ของ Ridge เพราะค่าที่ดีที่สุด (alpha=1) อยู่ขอบ grid และข้อ 10.6 ทำ ablation ของกลุ่ม Features เพื่อตอบว่าปัจจัยใดสำคัญเพียงใด

### 10.3 Metric ที่ใช้และเหตุผล

- **MAE (เกณฑ์หลัก):** ความคลาดเคลื่อนเฉลี่ยเป็นชั่วโมงจริง เข้าใจง่ายเชิงธุรกิจ และไม่ถูกเรื่องที่ปิดช้ามากดึงมากเท่า RMSE ซึ่งเหมาะกับ Target ที่เบ้ขวา
- **Median AE:** สะท้อนความคลาดเคลื่อนของเรื่องทั่วไป ไม่ถูกหางยาวบิดเบือน
- **RMSE:** ให้น้ำหนักกับความผิดพลาดรุนแรง ใช้ดูผลของเรื่องที่ปิดช้าผิดปกติ

### 10.4 Model selection and tuning

```python
# Model selection and tuning
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import ParameterGrid
from sklearn.metrics import mean_absolute_error, median_absolute_error, root_mean_squared_error
import time

tune_idx = X_train.sample(400_000, random_state=RANDOM_STATE).index   # สุ่มจาก train เท่านั้น
Xt, yt = X_train.loc[tune_idx], y_train.loc[tune_idx]

candidates = {
    'Ridge (cyclic)': ('cyclic', True, lambda **p: Ridge(**p), {'alpha': [1, 10, 100]}),
    'Tree (base)':    ('base', False, lambda **p: DecisionTreeRegressor(random_state=42, **p),
                       {'max_depth': [10], 'min_samples_leaf': [200]}),
    'Tree (tree)':    ('tree', False, lambda **p: DecisionTreeRegressor(random_state=42, **p),
                       {'max_depth': [8, 12, 16], 'min_samples_leaf': [100, 500]}),
    'RandomForest':   ('tree', False, lambda **p: RandomForestRegressor(n_jobs=-1, random_state=42, **p),
                       {'n_estimators': [100], 'max_depth': [12, 16], 'min_samples_leaf': [100]}),
}

rows = []
# baseline
dummy = DummyRegressor(strategy='median').fit(Xt, yt)
pred = dummy.predict(X_val)
rows.append(('Dummy median', {}, mean_absolute_error(y_val, pred),
             median_absolute_error(y_val, pred), root_mean_squared_error(y_val, pred), 0))

for name, (fset, scale, make_est, grid) in candidates.items():
    cols = feature_sets[fset]
    for params in ParameterGrid(grid):
        pipe = with_log_target(Pipeline([('prep', make_preprocessor(fset, scale)),
                                         ('model', make_est(**params))]))
        t0 = time.time()
        pipe.fit(Xt[cols], yt)
        pred = pipe.predict(X_val[cols])
        rows.append((name, params, mean_absolute_error(y_val, pred),
                     median_absolute_error(y_val, pred),
                     root_mean_squared_error(y_val, pred), time.time() - t0))

results = pd.DataFrame(rows, columns=['model', 'params', 'val_MAE', 'val_MedAE', 'val_RMSE', 'fit_sec'])
display(results.sort_values('val_MAE'))
```

### 11.3 วิเคราะห์ความผิดพลาดตามประเภทคำร้อง

```python

# 11.3 วิเคราะห์ความผิดพลาดตามประเภทคำร้อง

error_by_type = (
    results_df.groupby("ประเภทคำร้อง", dropna=False)
    .agg(
        requests=("เวลาปิดจริง (ชม.)", "size"),
        actual_mean=("เวลาปิดจริง (ชม.)", "mean"),
        predicted_mean=("เวลาทำนาย (ชม.)", "mean"),
        mae=("คลาดเคลื่อน (ชม.)", "mean")
    )
)

# 10 ประเภทคำร้องที่พบมากที่สุด
top10 = (
    error_by_type
    .sort_values("requests", ascending=False)
    .head(10)
)

print("=== ความผิดพลาดของ 10 ประเภทคำร้องที่พบมากที่สุด ===")

display(
    top10.rename(columns={
        "requests": "จำนวนคำร้อง",
        "actual_mean": "เวลาปิดจริงเฉลี่ย (ชม.)",
        "predicted_mean": "เวลาทำนายเฉลี่ย (ชม.)",
        "mae": "MAE (ชม.)"
    }).round(2)
)


# 10 ประเภทที่มี MAE สูงที่สุด
# พิจารณาเฉพาะประเภทที่มีคำร้องอย่างน้อย 1,000 รายการ
high_error = (
    error_by_type[error_by_type["requests"] >= 1000]
    .sort_values("mae", ascending=False)
    .head(10)
)

print("\n=== 10 ประเภทคำร้องที่มี MAE สูงที่สุด (อย่างน้อย 1,000 คำร้อง) ===")

display(
    high_error.rename(columns={
        "requests": "จำนวนคำร้อง",
        "actual_mean": "เวลาปิดจริงเฉลี่ย (ชม.)",
        "predicted_mean": "เวลาทำนายเฉลี่ย (ชม.)",
        "mae": "MAE (ชม.)"
    }).round(2)
)


# สรุปผล
most_common = error_by_type["requests"].idxmax()
worst_type = high_error["mae"].idxmax()

print("\n=== สรุปผล ===")
print(f"ประเภทคำร้องที่พบมากที่สุด: {most_common}")
print(f"ประเภทที่มี MAE สูงที่สุด (ในกลุ่ม ≥1,000 คำร้อง): {worst_type}")
print(f"MAE ของประเภทดังกล่าว: {high_error.loc[worst_type, 'mae']:,.2f} ชั่วโมง")
```

จากการวิเคราะห์ความผิดพลาดแยกตามประเภทคำร้อง พบว่าโมเดล Ridge Regression มีความคลาดเคลื่อนแตกต่างกันค่อนข้างมากในแต่ละประเภท

คำร้องที่เกี่ยวกับเสียงดังและการจอดรถ เช่น Noise - Residential และ Illegal Parking มีค่า MAE ประมาณ 2 ชั่วโมง ขณะที่คำร้องเกี่ยวกับสุขลักษณะ ระบบประปา และการซ่อมแซมอาคารมีค่า MAE สูงกว่าหลายร้อยชั่วโมง

เมื่อพิจารณาเฉพาะประเภทที่มีคำร้องอย่างน้อย 1,000 รายการ พบว่า Taxi Complaint มีค่า MAE สูงที่สุดที่ 2,565.67 ชั่วโมง รองลงมาคือ For Hire Vehicle Complaint ที่ 2,225.61 ชั่วโมง โดยทั้งสองประเภทมีระยะเวลาปิดจริงเฉลี่ยยาวมาก และโมเดลทำนายเวลาเฉลี่ยต่ำกว่าความจริง

### 11.4 Error Analysis by Resolution Duration

```python

# 11.4 Error Analysis by Resolution Duration

# แบ่งช่วงตามระยะเวลาปิดจริง (หน่วย: ชั่วโมง)
bins = [0, 24, 168, 720, np.inf]
labels = ["Within 24 hours", "1-7 days", "7-30 days", "Over 30 days"]

results_df["duration_group"] = pd.cut(
    results_df["เวลาปิดจริง (ชม.)"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# วิเคราะห์ความผิดพลาดในแต่ละช่วง
error_by_duration = (
    results_df.groupby("duration_group", observed=True)
    .agg(
        requests=("เวลาปิดจริง (ชม.)", "size"),
        actual_mean=("เวลาปิดจริง (ชม.)", "mean"),
        predicted_mean=("เวลาทำนาย (ชม.)", "mean"),
        mae=("คลาดเคลื่อน (ชม.)", "mean"),
        total_error=("คลาดเคลื่อน (ชม.)", "sum")
    )
)

# คิดสัดส่วนจำนวนคำร้องและความผิดพลาดรวม
error_by_duration["request_share"] = (
    error_by_duration["requests"] / len(results_df) * 100
)

error_by_duration["error_share"] = (
    error_by_duration["total_error"]
    / error_by_duration["total_error"].sum() * 100
)

print("=== ความผิดพลาดตามระยะเวลาปิดคำร้อง ===")

display(
    error_by_duration.drop(columns="total_error")
    .rename(columns={
        "requests": "จำนวนคำร้อง",
        "actual_mean": "เวลาปิดจริงเฉลี่ย (ชม.)",
        "predicted_mean": "เวลาทำนายเฉลี่ย (ชม.)",
        "mae": "MAE (ชม.)",
        "request_share": "สัดส่วนคำร้อง (%)",
        "error_share": "สัดส่วนความผิดพลาดรวม (%)"
    })
    .round(2)
)

# สรุปว่ากลุ่มใดมีผลต่อความผิดพลาดมากที่สุด
worst_group = error_by_duration["mae"].idxmax()
largest_share = error_by_duration["error_share"].idxmax()

print("\n=== สรุปผล ===")
print(f"ช่วงที่มี MAE สูงที่สุด: {worst_group}")
print(f"ช่วงที่สร้างความผิดพลาดรวมมากที่สุด: {largest_share}")
print(
    f"คิดเป็น {error_by_duration.loc[largest_share, 'error_share']:.2f}% "
    "ของความผิดพลาดสัมบูรณ์รวมทั้งหมด"
)

print("\nหมายเหตุ: ใช้เวลาปิดจริงแบ่งกลุ่มเพื่อวิเคราะห์ข้อผิดพลาด")
print("ไม่ได้ใช้ duration_group เป็นตัวแปรในการทำนาย")
```

จากการวิเคราะห์ความผิดพลาดตามระยะเวลาปิดคำร้อง พบว่าคำร้องที่ปิดภายใน 24 ชั่วโมงมีจำนวนมากที่สุด คิดเป็น 58.23% ของ Test Set แต่มีค่า MAE เพียง 10.43 ชั่วโมง

ในทางกลับกัน คำร้องที่ใช้เวลาปิดเกิน 30 วันมีเพียง 7.30% ของข้อมูล แต่มีค่า MAE สูงถึง 1,771.62 ชั่วโมง และคิดเป็น 78.79% ของความผิดพลาดสัมบูรณ์รวมทั้งหมด โดยโมเดลทำนายเวลาปิดเฉลี่ยเพียง 546.06 ชั่วโมง ขณะที่เวลาปิดจริงเฉลี่ยอยู่ที่ 2,282.46 ชั่วโมง

ผลการวิเคราะห์แสดงให้เห็นว่าคำร้องที่ใช้เวลาปิดนานมีผลต่อค่า MAE โดยรวมค่อนข้างมาก และเป็นข้อจำกัดสำคัญของ Ridge Regression ในงานนี้ อย่างไรก็ตาม คำร้องเหล่านี้ยังคงเป็นข้อมูลจริงที่มีความสำคัญ จึงไม่ควรตัดออกเพียงเพื่อให้ผลการประเมินโมเดลดีขึ้น

```python
cols = feature_sets["cyclic"]
_cl = split_info["closed_dt"]
strict_tr = train_mask & _cl.lt(TRAIN_END)
strict_va = val_mask & _cl.lt(VAL_END)
strict_fit = strict_tr | strict_va
drop_tr, drop_va = int((train_mask & ~strict_tr).sum()), int((val_mask & ~strict_va).sum())

strict_model = with_log_target(Pipeline([("prep", make_preprocessor("cyclic", True)), ("model", Ridge(alpha=1))]))
strict_model.fit(X.loc[strict_fit, cols], y.loc[strict_fit])
p_strict = np.maximum(strict_model.predict(X_test[cols]), 0)
dummy_strict = float(np.median(y.loc[strict_fit]))

cmp_df = pd.DataFrame({
    "Test MAE": [mean_absolute_error(y_true, y_pred), mean_absolute_error(y_true, p_strict),
                 mean_absolute_error(y_true, np.full_like(y_true, dummy_strict))],
    "Test MedAE": [median_absolute_error(y_true, y_pred), median_absolute_error(y_true, p_strict),
                   median_absolute_error(y_true, np.full_like(y_true, dummy_strict))],
    "Test RMSE": [root_mean_squared_error(y_true, y_pred), root_mean_squared_error(y_true, p_strict),
                  root_mean_squared_error(y_true, np.full_like(y_true, dummy_strict))],
}, index=["Ridge: กฎหลัก (Final Model ข้อ 11.1)", "Ridge: กฎเข้ม (ตัด label ที่ปิดหลังวันแบ่ง)", "Dummy median (กฎเข้ม)"])
display(cmp_df.round(2))

_mae_main, _mae_strict = cmp_df.iloc[0, 0], cmp_df.iloc[1, 0]
_diff = (_mae_strict - _mae_main) / _mae_main * 100
_gain_strict = (cmp_df.iloc[2, 0] - _mae_strict) / cmp_df.iloc[2, 0] * 100
print(f"แถวที่ถูกตัดเพิ่มในกฎเข้ม: Train {drop_tr:,}, Validation {drop_va:,}")

def _share(frame, col, val):
    return 100 * frame[col].eq(val).mean()
shift = pd.DataFrame({
    "median Target (ชม.)": [y_train.median(), y_val.median(), y_test.median()],
    "mean Target (ชม.)": [y_train.mean(), y_val.mean(), y_test.mean()],
    "สัดส่วน HEAT/HOT WATER (%)": [_share(X_train, "complaint_type", "HEAT/HOT WATER"),
                                   _share(X_val, "complaint_type", "HEAT/HOT WATER"),
                                   _share(X_test, "complaint_type", "HEAT/HOT WATER")],
    "สัดส่วน NYPD (%)": [_share(X_train, "agency", "NYPD"), _share(X_val, "agency", "NYPD"), _share(X_test, "agency", "NYPD")],
}, index=["Train", "Validation", "Test"])
display(shift.round(2))

_robust = ("ต่างกันไม่ถึง 5% จึงถือว่าข้อสรุปหลักไม่ไวต่อกฎ label availability"
           if abs(_diff) < 5 else
           "ต่างกันตั้งแต่ 5% ขึ้นไป ต้องรายงานเป็นข้อจำกัดสำคัญและพิจารณาใช้กฎเข้มเป็นผลหลัก")
_h_tr = shift.loc["Train", "สัดส่วน HEAT/HOT WATER (%)"]
_h_te = shift.loc["Test", "สัดส่วน HEAT/HOT WATER (%)"]
if _h_te > 1.2 * _h_tr:
    _shift_txt = ("ประเภทเรื่องนี้ (ซึ่งมี Target เฉลี่ยสูงกว่าเรื่องจราจร/เสียงดังมาก) พบบ่อยขึ้นอย่างชัดเจนในช่วง พ.ย.–ธ.ค. ซึ่งเป็นฤดูหนาว "
                  "เป็นการเปลี่ยนแปลงขององค์ประกอบข้อมูล (distribution shift) ที่ Train ซึ่งครอบคลุมเพียง ม.ค.–ส.ค. อธิบายได้ไม่เต็มที่")
else:
    _shift_txt = "ไม่พบการเปลี่ยนแปลงชัดเจนของสัดส่วนประเภทเรื่องหลัก จึงไม่มีหลักฐานของ distribution shift จากองค์ประกอบประเภทเรื่อง"
_medae_txt = (" และเป็นเหตุผลหนึ่งที่ MedAE ของ Ridge ไม่ดีกว่า Dummy ใน Test"
              if ridge_result["MedAE"] > dummy_result["MedAE"] and _h_te > 1.2 * _h_tr else "")
_medae_txt += "."
display(Markdown(f"""**ผล (ก):** กฎเข้มตัด label ที่ปิดหลังวันแบ่งเพิ่ม {drop_tr:,} แถวจาก Train และ {drop_va:,} แถวจาก Validation ได้ Test MAE {_mae_strict:.2f} ชม. เทียบกับกฎหลัก {_mae_main:.2f} ชม. ({_diff:+.2f}%) และยังดีกว่า Dummy median {_gain_strict:.1f}% → {_robust}

**ผล (ข):** มัธยฐาน Target ของ Test ({shift.loc['Test','median Target (ชม.)']:.2f} ชม.) เทียบกับ Validation ({shift.loc['Validation','median Target (ชม.)']:.2f} ชม.) และ Train ({shift.loc['Train','median Target (ชม.)']:.2f} ชม.); สัดส่วน HEAT/HOT WATER ของ Test {_h_te:.1f}% เทียบกับ Train {_h_tr:.1f}% → {_shift_txt}{_medae_txt}"""))
```

ข้อสังเกตเพิ่มเติม: ส่วนต่าง +4.89% ใกล้เกณฑ์ 5% ซึ่งเป็นเกณฑ์ที่กลุ่มกำหนดเอง จึงควรอ่านร่วมกับผลอื่น กฎเข้มให้ Test MAE 172.22 ชม. (กฎหลัก 164.19 ชม.) แต่ Ridge ยังดีกว่า Dummy ที่ใช้กฎเดียวกัน 18.0% (MAE 210.07 ชม.) และ MedAE ของ Ridge (6.11 ชม.) ยังสูงกว่า Dummy (5.26 ชม.) เช่นเดียวกับกฎหลัก สรุปได้ว่ากฎ label ส่งผลต่อระดับความคลาดเคลื่อนราว 5% แต่ไม่เปลี่ยนข้อสรุปว่าโมเดลดีกว่า Dummy ในแง่ MAE และยังอธิบายความแปรปรวนได้จำกัด ส่วนข้อความผล (ข) ที่ว่าสัดส่วน HEAT/HOT WATER ที่สูงขึ้นเป็นเหตุผลหนึ่งที่ MedAE ของ Ridge ไม่ดีกว่า Dummy เป็นข้อสันนิษฐานที่กลุ่มยังไม่ได้ทดสอบโดยตรง

### 12.1 เปรียบเทียบ Train / CV / Validation / Test

```python

# 12.1 เปรียบเทียบ Train / CV / Validation / Test
# ใช้ตัวอย่าง Train 150,000 คำร้อง เพื่อลดเวลาในการประมวลผล

import gc
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.dummy import DummyRegressor
from sklearn.model_selection import TimeSeriesSplit
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ใช้โมเดลและพารามิเตอร์ตามข้อ 10
def build_model(name):
    if name == "Ridge Regression":
        return with_log_target(Pipeline([
            ("prep", make_preprocessor("cyclic", True)),
            ("model", Ridge(alpha=1))
        ]))

    if name == "Decision Tree":
        return with_log_target(Pipeline([
            ("prep", make_preprocessor("tree", False)),
            ("model", DecisionTreeRegressor(
                max_depth=16,
                min_samples_leaf=500,
                random_state=42
            ))
        ]))

    if name == "Random Forest":
        return with_log_target(Pipeline([
            ("prep", make_preprocessor("tree", False)),
            ("model", RandomForestRegressor(
                n_estimators=100,
                max_depth=16,
                min_samples_leaf=100,
                random_state=42,
                n_jobs=-1
            ))
        ]))

    return DummyRegressor(strategy="median")


def evaluate(y, pred):
    pred = np.maximum(pred, 0)
    return {
        "MAE": mean_absolute_error(y, pred),
        "RMSE": np.sqrt(mean_squared_error(y, pred)),
        "R2": r2_score(y, pred)
    }


model_names = [
    "Ridge Regression",
    "Decision Tree",
    "Random Forest",
    "Dummy Median"
]

# ตรวจสอบวันสร้างคำร้อง
assert X_train.index.isin(clean_df.index).all()
assert not X_train.index.has_duplicates
assert not clean_df.index.has_duplicates

train_dates = pd.to_datetime(
    clean_df.loc[X_train.index, "created_date"],
    errors="coerce"
)

assert train_dates.notna().all()
assert (
    (train_dates >= "2025-01-01") &
    (train_dates < "2025-09-01")
).all(), "พบวันที่นอกช่วง Train"

# สุ่ม 150,000 คำร้องจาก Train และเรียงตามเวลา
rng = np.random.RandomState(42)
sample_idx = rng.choice(
    len(X_train),
    size=min(150_000, len(X_train)),
    replace=False
)

sample_idx = sample_idx[
    np.argsort(train_dates.iloc[sample_idx].to_numpy(), kind="stable")
]

X_sample = X_train.iloc[sample_idx].reset_index(drop=True)
y_sample = np.asarray(y_train)[sample_idx]

print(f"ข้อมูลตัวอย่าง Train: {len(X_sample):,} คำร้อง")

tscv = TimeSeriesSplit(n_splits=3)
comparison_rows = []
cv_rows = []

for name in model_names:
    print(f"\n=== {name} ===", flush=True)

    # เลือก Feature ให้ตรงกับโมเดล
    if name == "Dummy Median":
        Xtr = np.zeros((len(X_sample), 1))
        Xva = np.zeros((len(y_val), 1))
    else:
        fset = "cyclic" if name == "Ridge Regression" else "tree"
        cols = feature_sets[fset]
        Xtr = X_sample[cols]
        Xva = X_val[cols]

    # ประเมิน Train และ Validation
    model = build_model(name)
    model.fit(Xtr, y_sample)

    train_result = evaluate(y_sample, model.predict(Xtr))
    val_result = evaluate(y_val, model.predict(Xva))

    print(f"Train MAE: {train_result['MAE']:.2f} ชั่วโมง", flush=True)
    print(f"Validation MAE: {val_result['MAE']:.2f} ชั่วโมง", flush=True)

    del model
    gc.collect()

    # Cross-validation แบบเรียงตามเวลา 3 Fold
    fold_scores = []

    for fold, (tr_idx, va_idx) in enumerate(tscv.split(Xtr), 1):
        fold_model = build_model(name)

        if name == "Dummy Median":
            X_fold_tr = Xtr[tr_idx]
            X_fold_va = Xtr[va_idx]
        else:
            X_fold_tr = Xtr.iloc[tr_idx]
            X_fold_va = Xtr.iloc[va_idx]

        fold_model.fit(X_fold_tr, y_sample[tr_idx])

        result = evaluate(
            y_sample[va_idx],
            fold_model.predict(X_fold_va)
        )

        fold_scores.append(result)
        cv_rows.append({
            "Model": name,
            "Fold": fold,
            **result
        })

        print(
            f"Fold {fold}: MAE = {result['MAE']:.2f} ชั่วโมง",
            flush=True
        )

        del fold_model
        gc.collect()

    fold_df = pd.DataFrame(fold_scores)

    # Test ใช้ผลจาก Final Model ในข้อ 11.1
    if name in ["Ridge Regression", "Dummy Median"]:
        test_row = metrics.loc[
            metrics["Model"] == name
        ].iloc[0]

        test_result = {
            "MAE": test_row["MAE"],
            "RMSE": test_row["RMSE"],
            "R2": test_row["R2"]
        }
    else:
        test_result = {
            "MAE": np.nan,
            "RMSE": np.nan,
            "R2": np.nan
        }

    comparison_rows.append({
        "Model": name,
        "Train MAE": train_result["MAE"],
        "Validation MAE": val_result["MAE"],
        "Test MAE": test_result["MAE"],
        "Train RMSE": train_result["RMSE"],
        "Validation RMSE": val_result["RMSE"],
        "Test RMSE": test_result["RMSE"],
        "Train R2": train_result["R2"],
        "Validation R2": val_result["R2"],
        "Test R2": test_result["R2"],
        "CV MAE (mean ± SD)": (
            f"{fold_df['MAE'].mean():.2f} ± "
            f"{fold_df['MAE'].std(ddof=1):.2f}"
        ),
        "CV RMSE (mean ± SD)": (
            f"{fold_df['RMSE'].mean():.2f} ± "
            f"{fold_df['RMSE'].std(ddof=1):.2f}"
        ),
        "CV R2 (mean ± SD)": (
            f"{fold_df['R2'].mean():.4f} ± "
            f"{fold_df['R2'].std(ddof=1):.4f}"
        )
    })

    del Xtr, Xva
    gc.collect()


# ตารางเปรียบเทียบทุกโมเดล
comparison_all = pd.DataFrame(comparison_rows)

print("\n=== Model Performance Comparison ===")
display(comparison_all.round(2).fillna("Not evaluated"))

# รายละเอียดผล Cross-validation แต่ละ Fold
cv_detail_df = pd.DataFrame(cv_rows)

print("\n=== Cross-validation Results by Fold ===")
display(cv_detail_df.round(2))

print("\nหมายเหตุ:")
print("- Train และ CV ใช้ตัวอย่างจาก Train Set จำนวน 150,000 คำร้อง")
print("- ใช้ข้อมูลตัวอย่างเดียวกันในการเปรียบเทียบทั้ง 4 โมเดล")
print("- Validation ใช้ข้อมูล Validation Set เดิม")
print("- Test ของ Ridge และ Dummy ใช้ผลจากข้อ 11.1")
print("- Test ของ Decision Tree และ Random Forest ไม่ได้ประเมิน")
print("- การเลือก Final Model ยึดผล Model Selection ในข้อ 10")
```

จากการเปรียบเทียบโมเดลทั้ง 4 ตัว โดยใช้ข้อมูลตัวอย่าง Train จำนวน 150,000 คำร้อง พบว่า Ridge Regression (alpha=1) มีค่า Validation MAE ต่ำที่สุดที่ 149.46 ชั่วโมง รองลงมาคือ Decision Tree ที่ 157.59 ชั่วโมง และ Random Forest ที่ 158.51 ชั่วโมง

เมื่อทำ TimeSeriesSplit จำนวน 3 Fold พบว่า Ridge Regression มีค่า CV MAE เฉลี่ยต่ำที่สุดที่ 203.71 ± 31.49 ชั่วโมง ขณะที่ Random Forest และ Decision Tree มีค่า 210.99 ± 38.72 และ 231.94 ± 42.68 ชั่วโมง ตามลำดับ แสดงว่า Ridge ให้ผลดีกว่าโมเดลอื่นที่ทดลองตามเกณฑ์ MAE ทั้งบน Validation และ CV

สำหรับ Test Set ใช้ผลของ Final Model จากข้อ 11 ซึ่ง Ridge Regression มีค่า MAE 164.19 ชั่วโมง ต่ำกว่า Dummy Median ที่ 209.92 ชั่วโมง หรือดีขึ้น 21.79%

ทั้งนี้ ผล Train และ CV ใช้ข้อมูลตัวอย่าง 150,000 คำร้อง ส่วนผล Test มาจาก Final Model ที่ฝึกด้วย Train และ Validation รวมกัน จึงเป็นผลจากการฝึกคนละขนาดข้อมูล และไม่ควรนำตัวเลขมาเปรียบเทียบเสมือนเป็นโมเดลที่ฝึกจากข้อมูลชุดเดียวกัน

#### **สรุปเหตุผลในการเลือกโมเดล**

กลุ่มเราเลือก Ridge Regression (alpha=1) เพราะมีหลักฐานสนับสนุนจากหลายด้าน ได้แก่ ค่า Validation MAE ต่ำที่สุดในการทดลองข้อ 10 และ 12.1 ผล Cross-validation ที่มีค่า MAE เฉลี่ยต่ำกว่า Decision Tree และ Random Forest และผลบน Test Set ที่ลด MAE จาก Dummy Median ได้ 21.79%

อย่างไรก็ตาม ผล Error Analysis พบว่าคำร้องที่ใช้เวลาปิดเกิน 30 วันมีสัดส่วนความผิดพลาดสูงมาก จึงสรุปได้ว่า Ridge Regression เป็นโมเดลที่เหมาะสมที่สุดในบรรดาโมเดลที่ทดลอง แต่ยังมีข้อจำกัดในการทำนายคำร้องที่ใช้เวลาปิดนาน

**หมายเหตุ:** การเปรียบเทียบข้อ 12.1 ใช้ตัวอย่าง Train 150,000 คำร้อง ส่วนผล Test มาจาก Final Model ที่ฝึกด้วย Train และ Validation รวมกัน ขณะที่เวลาในการฝึกมาจากการทดลองข้อ 10 ซึ่งใช้ Train 400,000 คำร้อง

### **13.1 คำตอบของคำถามที่ตั้งไว้**

โครงงานนี้มีเป้าหมายเพื่อทำนายระยะเวลาปิดคำร้อง NYC 311 จากข้อมูลปี 2025 โดยใช้ข้อมูลที่ทราบขณะรับคำร้อง เช่น ประเภทคำร้อง หน่วยงานที่รับผิดชอบ เขตพื้นที่ และช่วงเวลาที่แจ้ง

จากการทดลองเปรียบเทียบ Ridge Regression, Decision Tree และ Random Forest พบว่า **Ridge Regression (alpha=1)** มีค่า Validation MAE ต่ำที่สุด จึงถูกเลือกเป็นโมเดลสุดท้าย

เมื่อนำโมเดลมาประเมินกับ Test Set จำนวน 628,199 คำร้อง พบว่ามีค่า MAE เท่ากับ **164.19 ชั่วโมง**, MedAE เท่ากับ **6.34 ชั่วโมง**, RMSE เท่ากับ **614.08 ชั่วโมง** และ R² เท่ากับ **0.254**

ส่วนคำถามว่าปัจจัยใดสำคัญเพียงใด ดูผล ablation ข้อ 10.6 และสรุปที่ท้ายข้อนี้ (ข้อ 13.4)

ผลการทดลองแสดงให้เห็นว่าสามารถใช้ข้อมูลเบื้องต้นของคำร้องเพื่อประมาณระยะเวลาปิดได้ แต่ความแม่นยำยังแตกต่างกันตามประเภทคำร้องและระยะเวลาที่ใช้ดำเนินการ

โครงงานใช้ข้อมูลคำร้องปี 2025 และแบ่ง Train, Validation และ Test ตามวันที่สร้างคำร้อง ดังนั้นผลการประเมินอาจเปลี่ยนแปลงเมื่อใช้กับข้อมูลปีอื่นหรือเมื่อรูปแบบคำร้องเปลี่ยนไป อีกทั้งข้อมูลที่นำมาศึกษาเป็นคำร้องที่ปิดแล้ว จึงยังไม่สามารถยืนยันประสิทธิภาพกับคำร้องที่ยังเปิดอยู่ได้

นอกจากนี้ ระยะเวลาปิดคำร้องในระบบไม่ได้หมายความว่าปัญหาทางกายภาพได้รับการแก้ไขเสร็จสมบูรณ์เสมอไป
